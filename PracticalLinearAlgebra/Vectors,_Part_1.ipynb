## Creating and Visualizing Vectors in NumPy

Dalam linear algebra, vector adalah daftar angka yang berurutan. Dua karakteristik penting dari vector adalah:

1. **Dimensionality**, yaitu jumlah elemen dalam vector.
2. **Orientation**, yaitu apakah vector berbentuk column vector, row vector, atau array 1D tanpa orientasi khusus.

Dalam Python, istilah dimensi bisa berbeda dengan matematika. Secara matematika, dimensionality berarti jumlah elemen vector. Di NumPy, dimensi array merujuk pada bentuk geometris penyimpanan array, misalnya `(3,)`, `(1, 3)`, atau `(3, 1)`.

```python
asList = [1,2,3]
asArray = np.array([1,2,3]) # 1D array
rowVec = np.array([ [1,2,3] ]) # row
colVec = np.array([ [1],[2],[3] ]) # column

print(f'asList: {np.shape(asList)}')
print(f'asArray: {asArray.shape}')
print(f'rowVec: {rowVec.shape}')
print(f'colVec: {colVec.shape}')
```

## Geometry of Vectors

Secara aljabar, vector adalah daftar angka berurutan. Secara geometri, vector adalah garis lurus yang memiliki panjang atau magnitude dan arah. Vector memiliki tail sebagai titik awal dan head sebagai titik akhir. Jika tail berada di origin, vector disebut berada pada standard position.

## Operations on Vectors

Operasi dasar pada vector meliputi penjumlahan, pengurangan, perkalian scalar, transpose, broadcasting, norm, dot product, Hadamard multiplication, outer product, dan orthogonal vector decomposition.

## Adding Two Vectors

Penjumlahan dua vector dilakukan dengan menjumlahkan elemen yang posisinya sama. Operasi ini hanya valid jika kedua vector memiliki dimensionality dan orientation yang sesuai.

```python
v = np.array([4,5,6])
w = np.array([10,20,30])
u = np.array([0,3,6,9])

vPlusW = v+w
print('v + w =', vPlusW)

try:
  uPlusW = u+w # error! dimensions mismatched!
  print(uPlusW)
except ValueError as e:
  print('Error karena dimensi tidak cocok:', e)
```

Jika row vector dijumlahkan dengan column vector, NumPy dapat melakukan broadcasting. Hasilnya bukan penjumlahan vector biasa, tetapi operasi broadcasting.

```python
v = np.array([[4,5,6]]) # row vector
w = np.array([[10,20,30]]).T # column vector

v+w
```

## Geometry of Vector Addition and Subtraction

Secara geometri, penjumlahan vector dilakukan dengan menempatkan tail satu vector pada head vector lain. Vector hasil penjumlahan bergerak dari tail pertama menuju head terakhir.

Pengurangan vector dilakukan dengan menempatkan dua vector pada tail yang sama, lalu vector selisih bergerak dari head vector negatif menuju head vector positif.

## Vector-Scalar Multiplication

Scalar adalah angka tunggal. Perkalian vector dengan scalar dilakukan dengan mengalikan setiap elemen vector dengan scalar tersebut.

Scalar dapat memperbesar, memperkecil, membuat vector menjadi nol, atau membalik arah vector jika scalar bernilai negatif.

```python
s = 2
a = [3,4,5] # as list
b = np.array(a) # as np array

print(a*s)
print(b*s)
```

Pada list Python, `a*s` berarti pengulangan list jika `s` adalah integer. Pada NumPy array, `b*s` berarti perkalian scalar-vector secara element-wise.

## Scalar-Vector Addition

Secara formal dalam linear algebra, penjumlahan scalar dengan vector tidak didefinisikan. Namun, Python memperbolehkannya dengan cara menambahkan scalar ke setiap elemen vector.

```python
s = 2
v = np.array([3,6])

s+v
```

## Transpose

Transpose mengubah column vector menjadi row vector, atau row vector menjadi column vector. Transpose menukar indeks row dan column. Jika transpose dilakukan dua kali, vector kembali ke orientasi awal.

## Vector Broadcasting in Python

Broadcasting adalah operasi komputasi modern yang mengulang operasi antara vector dengan elemen vector lain secara otomatis. Ini bukan prosedur standar dalam linear algebra tradisional, tetapi sering digunakan dalam numerical coding.

```python
v = np.array([[1,2,3]]).T # col vector
w = np.array([[10,20]]) # row vector

v + w # addition with broadcasting
```

## Vector Magnitude and Unit Vectors

Magnitude, geometric length, atau norm dari vector adalah jarak dari tail ke head. Rumusnya adalah akar dari jumlah kuadrat seluruh elemen vector.

Unit vector adalah vector dengan magnitude 1. Unit vector dapat dibuat dari vector nonzero dengan membagi vector tersebut menggunakan norm-nya.

```python
v = np.array([1,2,3,7,8,9])
v_dim = len(v) # math dimensionality
v_mag = np.linalg.norm(v) # math magnitude, length, or norm

print('Dimensionality:', v_dim)
print('Magnitude:', v_mag)
```

## The Vector Dot Product

Dot product adalah operasi penting dalam linear algebra. Dot product menghasilkan satu angka yang dihitung dengan mengalikan elemen-elemen yang bersesuaian, lalu menjumlahkan hasilnya. Dot product hanya valid untuk dua vector dengan dimensionality yang sama.

```python
v = np.array([1,2,3,4])
w = np.array([5,6,7,8])

np.dot(v,w)
```

Scalar multiplication pada salah satu vector akan menskalakan hasil dot product dengan jumlah yang sama.

```python
s = 10
np.dot(s*v,w)
```

## The Dot Product Is Distributive

Dot product bersifat distributive. Artinya, dot product dari sebuah vector dengan penjumlahan dua vector sama dengan penjumlahan dua dot product terpisah.

```python
a = np.array([ 0,1,2 ])
b = np.array([ 3,5,8 ])
c = np.array([ 13,21,34 ])

# the dot product is distributive
res1 = np.dot( a, b+c )
res2 = np.dot( a,b ) + np.dot( a,c )

print(res1)
print(res2)
```

## Geometry of the Dot Product

Secara geometri, dot product adalah hasil perkalian magnitude dua vector yang diskalakan oleh cosine sudut antara keduanya. Tanda dot product ditentukan oleh hubungan geometri antar vector. Vector orthogonal memiliki dot product bernilai nol.

## Hadamard Multiplication

Hadamard multiplication adalah perkalian element-wise. Dua vector harus memiliki dimensionality yang sama agar operasi ini valid.

```python
a = np.array([5,4,8,2])
b = np.array([1,0,.5])

try:
  a*b
except ValueError as e:
  print('Error karena dimensi berbeda:', e)

# perbaikan: samakan jumlah elemen vector
b_fix = np.array([1,0,.5,-1])
a*b_fix
```

## Outer Product

Outer product membuat matrix dari column vector dan row vector. Berbeda dari dot product yang menghasilkan scalar, outer product menghasilkan matrix. Dua vector pada outer product boleh memiliki dimensionality yang berbeda.

```python
v = np.array([[1],[2],[3]]) # column vector
w = np.array([[4,5]]) # row vector

np.dot(v,w)
```

```python
np.outer([1,2,3],[4,5])
```

## Orthogonal Vector Decomposition

Orthogonal vector decomposition memecah target vector menjadi dua komponen terhadap reference vector:

1. Komponen yang parallel terhadap reference vector.
2. Komponen yang orthogonal terhadap reference vector.

Formula komponen parallel:

\[
t_{\parallel r} = r \frac{t^T r}{r^T r}
\]

Formula komponen perpendicular:

\[
t_{\perp r} = t - t_{\parallel r}
\]

Dua komponen tersebut jika dijumlahkan akan menghasilkan target vector kembali.

## Exercise 2-3

Membuat fungsi yang menerima vector dan menghasilkan unit vector dengan arah yang sama. Jika input adalah zeros vector, unit vector tidak dapat dibuat karena norm bernilai 0.

```python
def unit_vector(v):
  v = np.array(v, dtype=float)
  mag = np.linalg.norm(v)
  if mag == 0:
    raise ValueError('Zeros vector tidak memiliki unit vector karena norm bernilai 0.')
  return v / mag

v = np.array([3,4])
print('unit vector:', unit_vector(v))
print('norm unit vector:', np.linalg.norm(unit_vector(v)))

try:
  print(unit_vector(np.array([0,0])))
except ValueError as e:
  print(e)
```

## Exercise 2-9

Mengimplementasikan orthogonal vector decomposition. Target vector `t` dipecah menjadi komponen parallel dan perpendicular terhadap reference vector `r`. Setelah itu, dicek bahwa kedua komponen menjumlah kembali menjadi `t` dan saling orthogonal.

```python
np.random.seed(2)
t = np.random.randn(2)
r = np.random.randn(2)

t_parallel = r * (np.dot(t,r) / np.dot(r,r))
t_perpendicular = t - t_parallel

plt.figure(figsize=(6,5))
plt.quiver(0,0,t[0],t[1],angles='xy',scale_units='xy',scale=1,label='t')
plt.quiver(0,0,r[0],r[1],angles='xy',scale_units='xy',scale=1,label='r')
plt.quiver(0,0,t_parallel[0],t_parallel[1],angles='xy',scale_units='xy',scale=1,label='t_parallel')
plt.quiver(0,0,t_perpendicular[0],t_perpendicular[1],angles='xy',scale_units='xy',scale=1,label='t_perpendicular')

all_points = np.vstack([t,r,t_parallel,t_perpendicular,[0,0]])
margin = 0.5
plt.xlim(all_points[:,0].min()-margin, all_points[:,0].max()+margin)
plt.ylim(all_points[:,1].min()-margin, all_points[:,1].max()+margin)
plt.grid(True)
plt.axhline(0)
plt.axvline(0)
plt.gca().set_aspect('equal', adjustable='box')
plt.legend()
plt.title('Orthogonal vector decomposition')
plt.show()

print('t:', t)
print('t_parallel + t_perpendicular:', t_parallel + t_perpendicular)
print('komponen menjumlah menjadi t:', np.allclose(t, t_parallel + t_perpendicular))
print('dot t_perpendicular dengan t_parallel:', np.dot(t_perpendicular,t_parallel))
print('orthogonal:', np.isclose(np.dot(t_perpendicular,t_parallel),0))
```

## Exercise 2-10

Mengimplementasikan bug pada denominator projection scalar, yaitu memakai `t^T t` sebagai denominator, bukan `r^T r`. Hasil dicek dengan sanity-check: komponen perpendicular yang benar harus orthogonal terhadap reference vector `r`.

```python
np.random.seed(5)
t = np.random.randn(2)
r = np.random.randn(2)

# kode benar
parallel_correct = r * (np.dot(t,r) / np.dot(r,r))
perp_correct = t - parallel_correct

# bug: denominator salah, menggunakan t^T t
parallel_bug = r * (np.dot(t,r) / np.dot(t,t))
perp_bug = t - parallel_bug

print('dot perpendicular benar dengan r:', np.dot(perp_correct,r))
print('dot perpendicular bug dengan r:', np.dot(perp_bug,r))
print('sanity-check benar:', np.isclose(np.dot(perp_correct,r),0))
print('sanity-check bug:', np.isclose(np.dot(perp_bug,r),0))
```
