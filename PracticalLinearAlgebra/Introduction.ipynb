## Apa Itu Linear Algebra dan Mengapa Perlu Dipelajari?

Linear algebra memiliki sejarah panjang dalam matematika. Matriks awalnya digunakan sebagai notasi ringkas untuk menyimpan kumpulan angka, seperti koordinat geometri dan sistem persamaan. ## Pembuktian Matematis vs Intuisi dari Coding

Tujuan mempelajari matematika adalah memahami matematika. Ada beberapa cara untuk membangun pemahaman:

Penjelasan, diagram, dan contoh numerik membantu membangun intuisi. Banyak contoh dilakukan dalam 2D atau 3D agar mudah divisualisasikan, tetapi prinsipnya dapat diterapkan ke dimensi yang lebih tinggi.

### Intuisi dari coding

## Contoh Soft Proof: Komutativitas Perkalian

Prinsip komutativitas perkalian menyatakan bahwa:

\[
a \times b = b \times a
\]

Kode berikut menghasilkan dua angka acak dan menguji apakah menukar urutan perkalian mengubah hasilnya.

```python
import numpy as np

a = np.random.randn()
b = np.random.randn()
a*b - b*a
```

Jika kode tersebut dijalankan beberapa kali dan hasilnya selalu `0.0`, maka kita memperoleh intuisi bahwa urutan perkalian tidak memengaruhi hasil. Namun, intuisi dari kode bukan pengganti pembuktian matematis formal. Tujuannya adalah membantu memahami konsep matematika tanpa harus langsung masuk ke sintaks dan argumen abstrak.
