## Membuat Fungsi Sendiri

Fungsi dibuat dengan keyword `def`, nama fungsi, input di dalam tanda kurung, dan tanda titik dua. Baris kode di dalam fungsi harus diindentasi. Output diberikan dengan keyword `return`.

```python
def add2numbers(n1,n2):
  total = n1+n2
  print(total)
  return total
```

```python
s = add2numbers(4,5)
print(s)
```

## Visualisasi

## Menerjemahkan Formula ke Kode

Formula matematika harus diterjemahkan ke kode dengan mendefinisikan variabel yang dibutuhkan. Misalnya, untuk fungsi `y = x^2`, variabel `x` perlu dibuat terlebih dahulu.

```python
x = np.arange(-4,5)
y = x**2
plt.plot(x,y,'s-')
plt.show()
```

Agar grafik lebih halus, gunakan resolusi lebih tinggi dengan `np.linspace()`.

```python
x = np.linspace(-4,4,42)
y = x**2
plt.plot(x,y,'s-')
plt.show()
```

Contoh berikut menerjemahkan fungsi sigmoid ke kode. Materi menekankan **soft coding**, yaitu menyimpan parameter sebagai variabel agar mudah diubah.

```python
x = np.linspace(-4,4,42)
alpha = 1.4
beta = 2
num = alpha # numerator
den = 1 + np.exp(-beta*x) # denominator
fx = num / den
plt.plot(x,fx,'s-')
plt.show()
```

## Control Flow

Control flow membuat kode dapat berubah perilaku berdasarkan kondisi tertentu.

### Comparator

Comparator menghasilkan nilai Boolean: `True` atau `False`.

```python
print( 4<5 ) # 1
print( 4>5 ) # 2
print( 4==5 ) # 3
```

### If Statement

`if` menjalankan kode hanya ketika kondisi bernilai benar.

```python
var = 4
if var==4:
  print(f'{var} equals 4!')
print("I'm outside the +for+ loop.")
```

```python
var = 4
if var==5:
  print(f'{var} equals 5!')
print("I'm outside the +for+ loop.")
```

### `elif` dan `else`

`elif` dipakai untuk kondisi tambahan, sedangkan `else` dijalankan jika semua kondisi sebelumnya salah.

```python
var = 4
if var==5:
  print('var is 5') # code 1
elif var>5:
  print('var > 5') # code 2
else:
  print('var < 5') # code 3
print('Outside the if-elif-else')
```

### Multiple Conditions

Kondisi dapat digabung dengan `and` dan `or`.

```python
if 4==4 and 4<10:
  print('Code example 1.')
if 4==5 and 4<10:
  print('Code example 2.')
if 4==5 or 4<10:
  print('Code example 3.')
```

### For Loop

`for` loop dipakai untuk mengulang kode. Fungsi `range()` sering digunakan sebagai iterable.

```python
for i in range(0,10):
  print(i+1)
```

```python
theList = [ 2,'hello',np.linspace(0,1,14) ]
for item in theList:
  print(item)
```

### Nested Control Statements

Control statement dapat ditumpuk di dalam control statement lain. Operator `%` adalah modulus, yaitu sisa hasil pembagian.

```python
powers = [0]*10
for i in range(len(powers)):
  if i%2==0 and i>0:
    print(f'{i} is an even number')
  if i>4:
    powers[i] = i**2
print(powers)
```
