## Vector Sets

Sekumpulan vektor disebut **set**. Set vektor biasanya ditulis dengan huruf kapital miring seperti `S` atau `V`.

Secara matematis:

$$
V = \{v_1, ..., v_n\}
$$

Set vektor dapat berisi jumlah vektor yang terbatas, tidak terbatas, atau bahkan kosong. Set kosong ditulis sebagai:

$$
V = \{\}
$$

## Linear Weighted Combination

**Linear weighted combination** adalah cara mencampur beberapa vektor dengan memberikan bobot tertentu pada tiap vektor.

Bentuk umumnya:

$$
w = \lambda_1v_1 + \lambda_2v_2 + ... + \lambda_nv_n
$$

Semua vektor harus memiliki dimensi yang sama agar operasi penjumlahan valid. Nilai $\lambda$ dapat berupa bilangan real apa pun, termasuk nol.

Contoh dari materi:

$$
\lambda_1 = 1, \lambda_2 = 2, \lambda_3 = -3
$$

$$
v_1 =
\begin{bmatrix}
4 \\
5 \\
1
\end{bmatrix},
v_2 =
\begin{bmatrix}
-4 \\
0 \\
-4
\end{bmatrix},
v_3 =
\begin{bmatrix}
1 \\
3 \\
2
\end{bmatrix}
$$

Hasilnya:

$$
w = \lambda_1v_1 + \lambda_2v_2 + \lambda_3v_3 =
\begin{bmatrix}
-7 \\
-4 \\
-13
\end{bmatrix}
$$

```python
l1 = 1
l2 = 2
l3 = -3

v1 = np.array([4,5,1])
v2 = np.array([-4,0,-4])
v3 = np.array([1,3,2])

l1*v1 + l2*v2 + l3*v3
```

Linear weighted combination digunakan pada banyak aplikasi, seperti model statistik, principal components analysis, dan artificial neural networks.

## Linear Independence

Satu set vektor disebut **linearly dependent** jika minimal satu vektor dalam set tersebut dapat ditulis sebagai linear weighted combination dari vektor lain dalam set yang sama.

Sebaliknya, satu set vektor disebut **linearly independent** jika tidak ada vektor yang dapat ditulis sebagai linear weighted combination dari vektor lain dalam set tersebut.

Contoh dari materi:

$$
V =
\left\{
\begin{bmatrix}
1 \\
3
\end{bmatrix},
\begin{bmatrix}
2 \\
7
\end{bmatrix}
\right\}
$$

Set ini linearly independent karena tidak ada satu vektor yang merupakan kelipatan linear dari vektor lainnya.

$$
S =
\left\{
\begin{bmatrix}
1 \\
3
\end{bmatrix},
\begin{bmatrix}
2 \\
6
\end{bmatrix}
\right\}
$$

Set ini linearly dependent karena salah satu vektor dapat dibuat dari kelipatan vektor lainnya.

Definisi formal linear dependence:

$$
0 = \lambda_1v_1 + \lambda_2v_2 + ... + \lambda_nv_n, \lambda \in \mathbb{R}
$$

Agar tidak menjadi solusi trivial, minimal satu nilai $\lambda$ harus tidak sama dengan nol.

## Independence dan Zeros Vector

Set vektor yang mengandung **zeros vector** otomatis menjadi linearly dependent. Alasannya, scalar multiple dari zeros vector tetap menghasilkan zeros vector.

## Subspace dan Span

**Subspace** adalah hasil dari semua kemungkinan linear weighted combination dari set vektor.

**Span** adalah mekanisme pembentukan subspace dari set vektor.

Untuk satu vektor:

$$
V =
\left\{
\begin{bmatrix}
1 \\
3
\end{bmatrix}
\right\}
$$

Span dari vektor tersebut adalah semua versi hasil perkalian skalar dari vektor itu. Bentuknya adalah subspace 1D berupa garis.

Untuk dua vektor di $\mathbb{R}^3$:

$$
V =
\left\{
\begin{bmatrix}
1 \\
0 \\
2
\end{bmatrix},
\begin{bmatrix}
-1 \\
1 \\
2
\end{bmatrix}
\right\}
$$

Dua vektor tersebut membentuk subspace 2D berupa bidang di ruang 3D.

Namun, dua vektor tidak selalu membentuk subspace 2D. Jika salah satu vektor sudah berada dalam span vektor lainnya, maka subspace yang dibentuk tetap 1D.

Contoh:

$$
V =
\left\{
\begin{bmatrix}
1 \\
1 \\
1
\end{bmatrix},
\begin{bmatrix}
2 \\
2 \\
2
\end{bmatrix}
\right\}
$$

Vektor kedua hanyalah kelipatan dari vektor pertama, sehingga subspace yang dibentuk tetap garis 1D.

## Perbedaan Subspace dan Span

Span dapat dipahami sebagai proses, sedangkan subspace adalah hasilnya. Set vektor melakukan span, dan hasil dari span tersebut adalah subspace.

## Basis

**Basis** adalah seperti alat ukur untuk mendeskripsikan sebuah ruang. Dalam linear algebra, basis adalah set vektor yang digunakan untuk mengukur atau mendeskripsikan data dalam suatu subspace.

Basis standar untuk 2D:

$$
S_2 =
\left\{
\begin{bmatrix}
1 \\
0
\end{bmatrix},
\begin{bmatrix}
0 \\
1
\end{bmatrix}
\right\}
$$

Basis standar untuk 3D:

$$
S_3 =
\left\{
\begin{bmatrix}
1 \\
0 \\
0
\end{bmatrix},
\begin{bmatrix}
0 \\
1 \\
0
\end{bmatrix},
\begin{bmatrix}
0 \\
0 \\
1
\end{bmatrix}
\right\}
$$

Basis standar memiliki vektor yang saling orthogonal dan memiliki panjang satu.

Contoh basis lain untuk $\mathbb{R}^2$:

$$
T =
\left\{
\begin{bmatrix}
3 \\
1
\end{bmatrix},
\begin{bmatrix}
-3 \\
1
\end{bmatrix}
\right\}
$$

Basis yang berbeda dapat mendeskripsikan titik yang sama dengan koordinat yang berbeda.

## Definisi Basis

Satu set vektor dapat menjadi basis untuk suatu subspace jika memenuhi dua syarat:

1. Set tersebut melakukan span terhadap subspace tersebut.
2. Set tersebut linearly independent.

Linear independence penting karena setiap vektor dalam subspace harus memiliki koordinat yang unik dalam basis tersebut.

## Ringkasan

Poin utama dari bab ini:

- Vector set adalah kumpulan vektor.
- Linear weighted combination berarti melakukan scalar multiplication dan penjumlahan vektor.
- Linear independence berarti tidak ada vektor dalam set yang dapat dibuat dari kombinasi linear vektor lain dalam set tersebut.
- Subspace adalah himpunan tak hingga dari semua kemungkinan linear weighted combination.
- Basis adalah set vektor yang melakukan span terhadap subspace dan linearly independent.

## Exercise 3-1

Tulis ulang kode linear weighted combination, tetapi simpan scalar dalam list dan vektor sebagai elemen dalam list. Gunakan for loop untuk melakukan operasi linear weighted combination. Inisialisasi output vector menggunakan `np.zeros()`. Konfirmasi bahwa hasilnya sama dengan kode sebelumnya.

```python
scalars = [1, 2, -3]
vectors = [
  np.array([4,5,1]),
  np.array([-4,0,-4]),
  np.array([1,3,2])
]

result_loop = np.zeros(3)

for i in range(len(scalars)):
  result_loop = result_loop + scalars[i] * vectors[i]

result_direct = l1*v1 + l2*v2 + l3*v3

print("Hasil dengan loop :", result_loop)
print("Hasil langsung    :", result_direct)
print("Sama?             :", np.array_equal(result_loop, result_direct))
```

## Exercise 3-2

Eksplorasi ketika menambahkan scalar dan vektor baru ke dalam list. Jika vektor baru berada di $\mathbb{R}^4$ sementara vektor lain berada di $\mathbb{R}^3$, maka operasi penjumlahan tidak valid karena dimensinya berbeda. Jika jumlah scalar lebih banyak daripada jumlah vektor, maka struktur input tidak seimbang.

```python
# Tambahan scalar dan vektor yang masih sama-sama di R^3
scalars_ok = [1, 2, -3, 4]
vectors_ok = [
  np.array([4,5,1]),
  np.array([-4,0,-4]),
  np.array([1,3,2]),
  np.array([2,2,2])
]

result_ok = np.zeros(3)

for i in range(len(scalars_ok)):
  result_ok = result_ok + scalars_ok[i] * vectors_ok[i]

print("Hasil ketika semua vektor berada di R^3:")
print(result_ok)
```

```python
# Kasus vektor baru berada di R^4
scalars_bad_dim = [1, 2, -3, 4]
vectors_bad_dim = [
  np.array([4,5,1]),
  np.array([-4,0,-4]),
  np.array([1,3,2]),
  np.array([2,2,2,2])
]

result_bad_dim = np.zeros(3)

try:
  for i in range(len(scalars_bad_dim)):
    result_bad_dim = result_bad_dim + scalars_bad_dim[i] * vectors_bad_dim[i]
except ValueError as error:
  print("Error karena dimensi vektor tidak cocok:")
  print(error)
```

```python
# Kasus jumlah scalar lebih banyak daripada jumlah vektor
scalars_too_many = [1, 2, -3, 4]
vectors_too_few = [
  np.array([4,5,1]),
  np.array([-4,0,-4]),
  np.array([1,3,2])
]

result_too_many = np.zeros(3)

try:
  for i in range(len(scalars_too_many)):
    result_too_many = result_too_many + scalars_too_many[i] * vectors_too_few[i]
except IndexError as error:
  print("Error karena jumlah scalar lebih banyak daripada jumlah vektor:")
  print(error)
```

## Exercise 3-3

Gambar titik acak dalam subspace. Pertama, gunakan satu vektor `[1, 3]`, lalu buat 100 scalar acak dari distribusi uniform antara `-4` dan `+4`. Kalikan scalar acak tersebut dengan vektor basis untuk menghasilkan 100 titik dalam subspace, lalu plot titik-titiknya.

```python
v = np.array([1,3])
scalars = np.random.uniform(-4,4,100)

points = np.zeros((100,2))

for i in range(100):
  points[i,:] = scalars[i] * v

plt.figure(figsize=(5,5))
plt.plot(points[:,0], points[:,1], 'ko')
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Subspace dari satu vektor [1, 3]")
plt.axis("equal")
plt.show()
```

Berikutnya, ulangi prosedur menggunakan dua vektor di $\mathbb{R}^3$:

$$
\begin{bmatrix}
3 \\
5 \\
1
\end{bmatrix}
\quad dan \quad
\begin{bmatrix}
0 \\
2 \\
2
\end{bmatrix}
$$

Karena ada dua vektor, dibutuhkan 100 x 2 scalar acak.

```python
import plotly.graph_objects as go

v1 = np.array([3,5,1])
v2 = np.array([0,2,2])

scalars = np.random.uniform(-4,4,(100,2))

points = np.zeros((100,3))

for i in range(100):
  points[i,:] = scalars[i,0]*v1 + scalars[i,1]*v2

fig = go.Figure(data=[go.Scatter3d(
  x=points[:,0],
  y=points[:,1],
  z=points[:,2],
  mode='markers'
)])

fig.update_layout(
  title="Subspace dari dua vektor di R^3",
  scene=dict(
    xaxis_title="x",
    yaxis_title="y",
    zaxis_title="z"
  )
)

fig.show()
```

Terakhir, ulangi kasus $\mathbb{R}^3$, tetapi vektor kedua dibuat menjadi setengah dari vektor pertama. Karena vektor kedua hanya kelipatan dari vektor pertama, titik-titik yang terbentuk berada pada subspace 1D.

```python
v1 = np.array([3,5,1])
v2 = 0.5 * v1

scalars = np.random.uniform(-4,4,(100,2))

points_dependent = np.zeros((100,3))

for i in range(100):
  points_dependent[i,:] = scalars[i,0]*v1 + scalars[i,1]*v2

fig = go.Figure(data=[go.Scatter3d(
  x=points_dependent[:,0],
  y=points_dependent[:,1],
  z=points_dependent[:,2],
  mode='markers'
)])

fig.update_layout(
  title="Subspace ketika vektor kedua = 1/2 vektor pertama",
  scene=dict(
    xaxis_title="x",
    yaxis_title="y",
    zaxis_title="z"
  )
)

fig.show()
```
