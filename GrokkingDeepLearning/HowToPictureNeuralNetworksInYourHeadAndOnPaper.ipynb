## Correlation summarization

Correlation summarization adalah kunci untuk lanjut ke neural networks yang lebih advanced tanpa kewalahan.

**Correlation summarization:** neural networks mencari direct dan indirect correlation antara input layer dan output layer, yang masing-masing ditentukan oleh input dataset dan output dataset.

Pada level tinggi, inilah yang dilakukan semua neural networks.

Karena neural network pada dasarnya adalah rangkaian matrices yang dihubungkan oleh layers, setiap weight matrix juga dapat dilihat dari sudut pandang correlation.

**Local correlation summarization:** setiap set of weights melakukan optimasi untuk mempelajari cara mengorelasikan input layer dengan apa yang dikatakan output layer seharusnya.

Jika hanya ada dua layers, yaitu input dan output, weight matrix mengetahui apa yang seharusnya terjadi pada output layer berdasarkan output dataset. Weight matrix mencari correlation antara input dataset dan output dataset karena keduanya tersimpan dalam input dan output layers.

Pada multiple layers, penjelasannya menjadi lebih detail.

**Global correlation summarization:** apa yang seharusnya terjadi pada earlier layer dapat ditentukan dari apa yang dikatakan later layer, lalu dikalikan dengan weights di antara keduanya. Dengan cara ini, later layers dapat memberi tahu earlier layers jenis signal yang dibutuhkan agar pada akhirnya correlation dengan output dapat ditemukan.

Cross-communication ini disebut backpropagation.

Ketika global correlation memberi tahu setiap layer apa yang seharusnya terjadi, local correlation dapat mengoptimalkan weights secara lokal. Final layer dapat mengatakan bahwa nilainya perlu sedikit lebih tinggi, lalu layer sebelumnya menerima pesan untuk mengirim signal yang lebih tinggi. Proses ini berlanjut ke layer-layer sebelumnya sampai setiap layer mengetahui neuron mana yang perlu lebih tinggi atau lebih rendah.

## Review: Vector-matrix multiplication

Vector-matrix multiplication melakukan beberapa weighted sums dari sebuah vector.

Matrix harus memiliki jumlah rows yang sama dengan jumlah values pada vector, sehingga setiap column dalam matrix dapat menjalankan weighted sum yang unik. Jika matrix memiliki empat columns, maka empat weighted sums akan dibuat.

Pada gambar pertama, satu datapoint dari dataset streetlight dipilih. `layer_0` diisi dengan nilai yang sesuai.

Pada gambar kedua, empat weighted sums dari `layer_0` dilakukan oleh `weights_0_1`. Empat nilai tersebut disimpan di empat posisi `layer_1`, lalu melewati fungsi `relu` yang mengubah nilai negatif menjadi 0.

Pada gambar ketiga, langkah akhir melakukan weighted average dari `layer_1` menggunakan vector-matrix multiplication lagi. Hasil akhirnya adalah `0.9`, yaitu final prediction network.

```python
import numpy as np

def relu(x):
  return (x > 0) * x

layer_0 = np.array([[1, 0, 1]])

weights_0_1 = np.array([
  [0.5, 0.2, -0.1, 0.9],
  [0.1, 0.1, 0.5, 0.1],
  [0.0, 0.0, 0.1, 0.0]
])

weights_1_2 = np.array([
  [0.2],
  [0.3],
  [0.4],
  [0.8]
])

layer_1 = relu(layer_0.dot(weights_0_1))
layer_2 = layer_1.dot(weights_1_2)

print("layer_0:", layer_0)
print("layer_1:", layer_1)
print("layer_2:", layer_2)
```

Kode tersebut menunjukkan bentuk forward propagation yang sama dengan visualisasi: `layer_0` dikalikan dengan `weights_0_1`, hasilnya melewati `relu`, lalu dikalikan lagi dengan `weights_1_2`.

## Visualizing using letters instead of pictures

Gambar dan penjelasan detail sebenarnya dapat diringkas menjadi algebra sederhana.

Matrix dapat divisualisasikan dengan huruf kapital. Contohnya, `W` dapat digunakan untuk weights. Karena ada beberapa weight matrices, indeks seperti `W0` dan `W1` digunakan untuk membedakannya.

Vector dapat divisualisasikan dengan huruf kecil. Contohnya, `l` digunakan untuk layers, sehingga layer dapat ditulis sebagai `l0`, `l1`, dan `l2`.

## Linking the variables

Huruf-huruf tersebut dapat digabungkan untuk menunjukkan functions dan operations.

Vector-matrix multiplication ditulis dengan menaruh dua simbol berdampingan. Contohnya:

`l0W0`

Artinya: ambil vector layer 0 dan lakukan vector-matrix multiplication dengan weight matrix 0.

`l1W1`

Artinya: ambil vector layer 1 dan lakukan vector-matrix multiplication dengan weight matrix 1.

Fungsi seperti `relu` dapat dimasukkan dengan notasi yang mirip Python.

`l1 = relu(l0W0)`

Artinya: untuk membuat layer 1, ambil layer 0, lakukan vector-matrix multiplication dengan weight matrix 0, lalu jalankan `relu` pada hasilnya.

`l2 = l1W1`

Artinya: untuk membuat layer 2, ambil layer 1 dan lakukan vector-matrix multiplication dengan weight matrix 1.

Karena `l2` memakai `l1`, seluruh forward propagation dapat dirangkai menjadi satu formula:

`l2 = relu(l0W0)W1`

Formula ini mengasumsikan bahwa vectors dan matrices memiliki dimensions yang benar.

```python
import numpy as np

def relu(x):
  return (x > 0) * x

l0 = np.array([[1, 0, 1]])

W0 = np.array([
  [0.5, 0.2, -0.1, 0.9],
  [0.1, 0.1, 0.5, 0.1],
  [0.0, 0.0, 0.1, 0.0]
])

W1 = np.array([
  [0.2],
  [0.3],
  [0.4],
  [0.8]
])

l1 = relu(l0.dot(W0))
l2 = l1.dot(W1)

print("l1:", l1)
print("l2:", l2)

l2_one_line = relu(l0.dot(W0)).dot(W1)
print("l2 one-line:", l2_one_line)
```

## Everything side by side

Forward propagation yang sama dapat dilihat sebagai visualisasi, algebra formula, dan Python code.

Formula algebra:

`l2 = relu(l0W0)W1`

Kode Python:

`layer_2 = relu(layer_0.dot(weights_0_1)).dot(weights_1_2)`

Keduanya merepresentasikan forward propagation yang sama: input masuk ke hidden layer melalui matrix weights pertama, melewati `relu`, lalu masuk ke output layer melalui matrix weights kedua.

```python
import numpy as np

def relu(x):
  return (x > 0) * x

layer_0 = np.array([[1, 0, 1]])

weights_0_1 = np.array([
  [0.5, 0.2, -0.1, 0.9],
  [0.1, 0.1, 0.5, 0.1],
  [0.0, 0.0, 0.1, 0.0]
])

weights_1_2 = np.array([
  [0.2],
  [0.3],
  [0.4],
  [0.8]
])

layer_2 = relu(layer_0.dot(weights_0_1)).dot(weights_1_2)

print(layer_2)
```
