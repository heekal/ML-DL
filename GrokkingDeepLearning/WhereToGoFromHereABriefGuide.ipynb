## Step 1: Start learning PyTorch

Framework deep learning yang dibuat pada chapter 13 paling mirip dengan PyTorch.

Sampai titik ini, pembelajaran dilakukan menggunakan NumPy, lalu memakai toolkit deep learning kecil yang dibuat sendiri. Namun, untuk eksperimen berikutnya, kecuali saat belajar architecture baru, sebaiknya gunakan framework deep learning sungguhan.

Framework akan lebih sedikit bug, jauh lebih cepat, dan memungkinkan pengguna mempelajari serta memakai kode orang lain.

PyTorch disarankan karena terasa familiar bagi pengguna yang datang dari latar belakang NumPy. Selain itu, framework kecil yang dibuat pada chapter 13 memang didesain agar mirip dengan API PyTorch, sehingga transisinya lebih mudah.

Cara belajar PyTorch yang disarankan adalah mengambil course deep learning yang mengajarkan konsep deep learning menggunakan framework tersebut. Dengan begitu, konsep yang sudah dipelajari akan diulang, sambil mempelajari tempat konsep itu berada dalam API PyTorch.

Sumber yang disebutkan pada chapter ini meliputi Udacity Deep Learning Nanodegree, fast.ai, tutorial PyTorch, dan repository contoh PyTorch.

## Step 2: Start another deep learning course

Satu buku atau course saja tidak cukup untuk seluruh pendidikan deep learning.

Konsep yang sama perlu dipelajari berulang dari berbagai sudut pandang agar benar-benar dipahami. Bahkan jika semua konsep sudah dibahas dalam satu sumber, mendengarnya dari perspektif berbeda tetap penting.

Chapter ini menyarankan mencari online courses dari universitas besar atau AI labs, seperti Stanford, MIT, Oxford, Montreal, NYU, dan lainnya. Tonton videonya, kerjakan latihannya, dan ulangi konsep yang sama sampai fundamental terasa natural.

Tujuannya adalah membuat dasar-dasar deep learning menjadi second nature.

## Step 3: Grab a mathy deep learning textbook

Pengetahuan matematika dapat diperkuat dengan membalik proses belajar: mulai dari kode deep learning, lalu kembali ke matematika yang diimplementasikan oleh kode tersebut.

Chapter ini menyebut bahwa pengalaman membuat kode NumPy, lalu mencari tahu matematika di baliknya, dapat membantu memahami aljabar, kalkulus, dan statistik yang relevan dengan deep learning.

Buku matematika deep learning yang disarankan adalah *Deep Learning* oleh Ian Goodfellow, Yoshua Bengio, dan Aaron Courville. Buku tersebut dianggap sebagai langkah berikutnya setelah buku ini karena lebih kuat pada sisi matematika, tetapi tetap relevan untuk pembaca yang sudah memahami konsep dasarnya.

## Step 4: Start a blog, and teach deep learning

Mengajar deep learning melalui blog disebut sebagai salah satu cara terbaik untuk memperkuat pemahaman dan karier.

Mengajar memaksa seseorang menjelaskan konsep sesederhana mungkin. Ketika menulis untuk publik, penulis juga terdorong membuat penjelasan yang jelas, lengkap, dan mudah diikuti.

Chapter ini menyarankan untuk tidak hanya menulis ringkasan konsep. Ringkasan dianggap membosankan dan kurang bermanfaat. Sebaliknya, tulislah tutorial. Setiap blog post sebaiknya berisi neural network yang belajar melakukan sesuatu, dapat diunduh, dapat dijalankan, dan dijelaskan baris demi baris.

Jika ingin melamar pekerjaan, program master, atau PhD di AI, chapter ini menyarankan memilih researcher yang ingin diajak bekerja, lalu menulis tutorial tentang karya mereka. Ini menunjukkan pemahaman terhadap konsep yang mereka kerjakan dan dapat menjadi cara lebih baik daripada cold email.

## Step 5: Twitter

Banyak percakapan AI terjadi di Twitter.

Chapter ini menyebut Twitter sebagai tempat untuk mengikuti peneliti AI, mengetahui paper terbaru, dan menjadi bagian dari percakapan. Cara memulainya adalah mengikuti beberapa AI researchers yang dikagumi, lalu mengikuti orang-orang yang mereka ikuti.

Tujuannya bukan hanya mengikuti update, tetapi juga masuk ke percakapan yang sedang berjalan di komunitas AI.

## Step 6: Implement academic papers

Gabungan Twitter dan blog dapat digunakan untuk membuat tutorial berdasarkan academic papers.

Ketika menemukan paper yang menarik dan tidak membutuhkan jumlah GPU yang terlalu besar, chapter ini menyarankan untuk menulis tutorial tentang paper tersebut. Prosesnya memaksa pembaca membaca paper, memahami matematika, dan mencoba tuning yang juga dilakukan peneliti aslinya.

Latihan ini sangat berguna bagi orang yang tertarik pada abstract research. Dengan mengimplementasikan paper, seseorang dapat mulai memahami cara kerja riset, lalu mungkin menemukan cara untuk memperbaiki ide yang sudah ada.

## Step 7: Acquire access to a GPU (or many)

GPU mempercepat eksperimen, dan eksperimen yang lebih cepat membuat proses belajar lebih cepat.

Chapter ini menyebut bahwa GPU dapat memberi training 10 sampai 100 kali lebih cepat. Dampaknya adalah seseorang dapat mencoba ide baik dan buruk jauh lebih cepat.

Akses GPU bisa diperoleh dengan membeli GPU NVIDIA atau menggunakan GPU gratis di Google Colab. Chapter ini juga menyebut bahwa NVIDIA kadang memberikan akses gratis untuk student pada kompetisi AI tertentu.

## Step 8: Get paid to practice

Semakin banyak waktu yang tersedia untuk melakukan deep learning, semakin cepat seseorang belajar.

Chapter ini menyarankan mencari pekerjaan yang memungkinkan seseorang terus belajar selama jam kerja. Contohnya adalah data scientist, data engineer, research engineer, atau freelance consultant di bidang statistik.

Blog disebut penting untuk mendapatkan pekerjaan seperti ini. Untuk pekerjaan yang diinginkan, tulislah setidaknya dua blog post yang menunjukkan kemampuan melakukan hal yang dicari oleh pekerjaan tersebut. Kandidat yang ideal adalah orang yang sudah menunjukkan bahwa ia bisa mengerjakan pekerjaan itu.

## Step 9: Join an open source project

Salah satu cara terbaik untuk networking dan membangun karier di AI adalah menjadi core developer pada open source project.

Chapter ini menyarankan memilih framework deep learning yang disukai, lalu mulai berkontribusi. Langkah awal bisa berupa menambahkan unit tests dan dokumentasi, kemudian memperbaiki bug, lalu masuk ke proyek yang lebih besar.

Contoh project yang disebutkan adalah PyTorch, TensorFlow, Keras, dan OpenMined. Kontribusi open source memungkinkan seseorang berinteraksi dengan peneliti dan developer dari lab besar melalui review pull request dan pengembangan project.

## Step 10: Develop your local community

Belajar deep learning dapat lebih mudah ketika dilakukan bersama komunitas lokal.

Chapter ini menekankan pentingnya berada di tempat yang menyenangkan, bersama orang yang juga tertarik pada hal yang sama. Ketika konsep sulit atau bug sulit diperbaiki, komunitas dapat membantu seseorang bertahan lebih lama dan berkembang lebih cepat.

Membangun komunitas lokal bukan hanya tentang belajar, tetapi juga tentang menciptakan lingkungan yang membuat proses belajar lebih konsisten dan menyenangkan.
