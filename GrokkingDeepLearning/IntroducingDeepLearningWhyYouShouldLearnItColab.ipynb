## Welcome to Grokking Deep Learning

Deep learning berada pada persimpangan antara machine learning dan artificial intelligence. Bidang ini membawa perubahan besar bagi masyarakat dan industri. Metode yang dibahas dalam buku ini sedang mengubah banyak hal di sekitar kita, mulai dari optimasi mesin mobil hingga penentuan konten yang dilihat di media sosial. Deep learning ada di mana-mana, kuat, dan juga menyenangkan untuk dipelajari.

## Why you should learn deep learning

Deep learning adalah alat yang kuat untuk otomatisasi kecerdasan secara bertahap.

Sejak awal kehidupan manusia, manusia terus membangun alat yang semakin baik untuk memahami dan mengendalikan lingkungan di sekitarnya. Deep learning adalah bagian terbaru dari cerita inovasi tersebut.

Hal yang membuat deep learning menarik adalah bidang ini lebih merupakan inovasi mental dibandingkan inovasi mekanis. Sama seperti machine learning, deep learning berusaha mengotomatisasi kecerdasan sedikit demi sedikit. Dalam beberapa tahun terakhir, bidang ini mencapai kemajuan besar dalam computer vision, speech recognition, machine translation, dan banyak tugas lainnya.

Hal tersebut menjadi luar biasa karena deep learning tampaknya menggunakan algoritma yang sebagian besar sama, yaitu neural networks yang terinspirasi dari otak, untuk mencapai hasil di banyak bidang berbeda. Walaupun deep learning masih terus berkembang dan masih memiliki banyak tantangan, perkembangan terbaru di bidang ini menimbulkan antusiasme besar. Mungkin kita tidak hanya menemukan alat yang hebat, tetapi juga jendela untuk memahami pikiran kita sendiri.

## Deep learning has the potential for significant automation of skilled labor.

Ada banyak hype tentang dampak deep learning jika perkembangan saat ini terus berlanjut dengan berbagai kemungkinan kecepatan. Walaupun banyak prediksi terlalu berlebihan, ada satu hal yang layak dipertimbangkan, yaitu perpindahan pekerjaan.

Klaim ini penting karena sekalipun inovasi deep learning berhenti hari ini, dampaknya terhadap tenaga kerja terampil di seluruh dunia sudah sangat besar. Operator call center, sopir taksi, dan analis bisnis tingkat rendah adalah contoh pekerjaan yang dapat terdampak karena deep learning dapat menyediakan alternatif berbiaya rendah.

Untungnya, ekonomi tidak berubah secara instan. Namun, dalam banyak hal, kita sudah melewati titik yang perlu diperhatikan karena kekuatan teknologi ini saat ini. Harapannya, buku ini dapat membantu pembaca berpindah dari industri yang mungkin terdampak menuju industri yang tumbuh dan memiliki peluang besar, yaitu deep learning.

## It’s fun and creative. You’ll discover much about what it is to be human by trying to simulate intelligence and creativity.

Deep learning menarik karena berada di antara manusia dan mesin. Memahami apa arti berpikir, bernalar, dan mencipta adalah pengalaman yang mencerahkan dan menginspirasi.

Bayangkan ada dataset berisi semua lukisan yang pernah dibuat, lalu dataset tersebut digunakan untuk mengajari mesin melukis seperti Monet. Hal itu mungkin dilakukan, dan sangat menarik untuk melihat cara kerjanya.

Buku ini memiliki hambatan masuk yang rendah.

## It will help you understand what’s inside a framework (Torch, TensorFlow, and so on).

Ada dua kelompok besar materi pembelajaran deep learning. Kelompok pertama berfokus pada cara menggunakan framework dan library populer seperti Torch, TensorFlow, Keras, dan lainnya. Kelompok kedua berfokus pada pengajaran deep learning itu sendiri, yaitu ilmu yang berada di balik framework tersebut.

Pada akhirnya, keduanya penting untuk dipelajari. Ibarat ingin menjadi pembalap NASCAR, kamu perlu memahami model mobil yang kamu pakai, yaitu framework, dan juga memahami cara mengemudi, yaitu ilmu atau keterampilannya.

Namun, mempelajari framework saja ibarat mempelajari kelebihan dan kekurangan Generation 6 Chevrolet SS sebelum memahami apa itu stick shift. Buku ini bertujuan mengajarkan apa itu deep learning agar kamu siap mempelajari framework setelahnya.

## What you need to get started

Yang dibutuhkan untuk mulai belajar adalah Jupyter Notebook dan library Python NumPy.

Jupyter Notebook sangat berguna untuk belajar deep learning karena memungkinkan kita menghentikan network saat training dan membongkar setiap bagiannya untuk melihat bentuknya.

NumPy digunakan karena buku ini hanya memakai satu library matrix. Dengan cara ini, pembaca dapat memahami bagaimana semuanya bekerja, bukan hanya cara memanggil framework. Buku ini mengajarkan deep learning dari awal sampai akhir.

Instruksi instalasi untuk Jupyter tersedia di `http://jupyter.org`, sedangkan NumPy tersedia di `http://numpy.org`. Contoh dalam buku dibuat dengan Python 2.7, tetapi juga sudah diuji untuk Python 3. Untuk instalasi yang mudah, buku ini juga merekomendasikan Anaconda melalui `https://docs.continuum.io/anaconda/install`.
