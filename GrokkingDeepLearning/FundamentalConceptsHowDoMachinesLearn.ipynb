## What is deep learning?

Deep learning adalah bagian dari metode machine learning.

Deep learning adalah bagian dari machine learning, yaitu bidang yang mempelajari dan mengembangkan mesin yang dapat belajar. Dalam beberapa konteks, tujuan akhirnya dapat mengarah pada pencapaian general artificial intelligence.

Di industri, deep learning digunakan untuk menyelesaikan tugas praktis di berbagai bidang, seperti computer vision untuk gambar, natural language processing untuk teks, dan automatic speech recognition untuk audio.

Singkatnya, deep learning adalah bagian dari kumpulan metode dalam machine learning. Deep learning terutama menggunakan artificial neural networks, yaitu kelas algoritma yang secara longgar terinspirasi dari otak manusia.

Gambar tersebut menunjukkan bahwa tidak semua deep learning berfokus pada pencarian generalized artificial intelligence, seperti mesin sadar diri yang sering muncul di film. Banyak aplikasi teknologi ini digunakan untuk menyelesaikan berbagai masalah di industri.

Buku ini berfokus pada dasar-dasar deep learning yang berada di balik riset mutakhir dan kebutuhan industri, sehingga pembaca dapat mempersiapkan diri untuk keduanya.

## What is machine learning?

Machine learning adalah bidang studi yang memberi komputer kemampuan untuk belajar tanpa diprogram secara eksplisit.

Karena deep learning adalah bagian dari machine learning, pertanyaan berikutnya adalah: apa itu machine learning? Secara umum, machine learning adalah subbidang computer science tempat mesin belajar menjalankan tugas yang sebelumnya tidak diprogram secara eksplisit.

Singkatnya, mesin mengamati sebuah pola dan mencoba menirunya, baik secara langsung maupun tidak langsung.

Peniruan langsung dan tidak langsung ini sejajar dengan dua jenis utama machine learning, yaitu supervised dan unsupervised.

Supervised machine learning adalah peniruan langsung dari pola antara dua dataset. Metode ini selalu mencoba mengambil dataset input dan mengubahnya menjadi dataset output. Kemampuan ini sangat kuat dan berguna.

Contohnya:

- Menggunakan pixel dari gambar untuk mendeteksi ada atau tidaknya kucing.
- Menggunakan film yang disukai untuk memprediksi film lain yang mungkin disukai.
- Menggunakan kata-kata seseorang untuk memprediksi apakah ia senang atau sedih.
- Menggunakan data sensor cuaca untuk memprediksi kemungkinan hujan.
- Menggunakan sensor mesin mobil untuk memprediksi pengaturan tuning yang optimal.
- Menggunakan data berita untuk memprediksi harga saham besok.
- Menggunakan angka input untuk memprediksi angka yang nilainya dua kali lebih besar.
- Menggunakan file audio mentah untuk memprediksi transkrip audio tersebut.

Semua contoh tersebut adalah tugas supervised machine learning. Dalam semua kasus, algoritma machine learning mencoba meniru pola antara dua dataset sehingga satu dataset dapat digunakan untuk memprediksi dataset lainnya.

## Supervised machine learning

Supervised learning mengubah satu dataset menjadi dataset lain.

Misalnya, ada dataset bernama Monday Stock Prices yang mencatat harga setiap saham pada setiap hari Senin selama 10 tahun terakhir. Lalu ada dataset kedua bernama Tuesday Stock Prices yang mencatat data pada periode waktu yang sama. Algoritma supervised learning dapat mencoba menggunakan data hari Senin untuk memprediksi data hari Selasa.

Jika algoritma supervised machine learning berhasil dilatih menggunakan data Senin dan Selasa selama 10 tahun, maka harga saham pada hari Selasa di masa depan dapat diprediksi berdasarkan harga saham pada hari Senin tepat sebelumnya.

Supervised machine learning adalah bagian utama dari applied artificial intelligence, atau narrow AI. Metode ini berguna untuk mengambil sesuatu yang sudah diketahui sebagai input, lalu mengubahnya dengan cepat menjadi sesuatu yang ingin diketahui.

Hal tersebut membuat algoritma supervised machine learning dapat memperluas kecerdasan dan kemampuan manusia dalam banyak cara.

Sebagian besar pekerjaan yang menggunakan machine learning menghasilkan pelatihan supervised classifier. Bahkan unsupervised machine learning biasanya dilakukan untuk membantu pengembangan algoritma supervised machine learning yang akurat.

Untuk bagian berikutnya dalam buku ini, algoritma yang dibuat akan mengambil data input yang dapat diamati, direkam, dan diketahui, lalu mengubahnya menjadi data output berharga yang membutuhkan analisis logis. Inilah kekuatan supervised machine learning.

## Unsupervised machine learning

Unsupervised learning mengelompokkan data.

Unsupervised learning memiliki kesamaan dengan supervised learning, yaitu sama-sama mengubah satu dataset menjadi dataset lain. Namun, dataset hasil transformasinya belum diketahui atau belum dipahami sebelumnya.

Tidak seperti supervised learning, tidak ada “jawaban benar” yang ingin dibuat agar model menirunya. Kita hanya meminta algoritma unsupervised untuk menemukan pola dalam data dan memberi tahu pola tersebut.

Contohnya, clustering dataset ke dalam beberapa kelompok adalah salah satu bentuk unsupervised learning. Clustering mengubah rangkaian datapoint menjadi rangkaian label cluster. Jika algoritma menemukan 10 cluster, label yang umum digunakan adalah angka 1 sampai 10. Setiap datapoint akan diberi angka berdasarkan cluster tempat datapoint tersebut berada.

Dataset berubah dari kumpulan datapoint menjadi kumpulan label. Label tersebut berupa angka karena algoritma tidak memberi tahu nama cluster tersebut. Algoritma hanya menemukan bahwa ada struktur atau kelompok dalam data.

Ide clustering ini dapat digunakan sebagai pegangan untuk memahami unsupervised learning. Walaupun ada banyak bentuk unsupervised learning, semua bentuk unsupervised learning dapat dilihat sebagai bentuk clustering.

Pada contoh tersebut, algoritma tidak memberi tahu nama cluster. Namun, pengelompokan katanya dapat dipahami sebagai: 1 berarti cute dan 2 berarti delicious. Bagian berikutnya dalam buku akan menjelaskan bagaimana bentuk lain dari unsupervised learning juga dapat dilihat sebagai clustering dan mengapa cluster ini berguna untuk supervised learning.

## Parametric vs. nonparametric learning

Secara sederhana: trial-and-error learning dibandingkan counting and probability.

Bagian sebelumnya membagi algoritma machine learning menjadi dua kelompok: supervised dan unsupervised. Sekarang algoritma yang sama dibagi lagi dengan cara lain, yaitu parametric dan nonparametric.

Dengan begitu, machine learning memiliki dua pengaturan: supervised atau unsupervised, dan parametric atau nonparametric.

Ada empat jenis algoritma yang dapat dipilih. Sebuah algoritma dapat berupa unsupervised atau supervised, dan juga dapat berupa parametric atau nonparametric.

Bagian supervision membahas jenis pola yang dipelajari. Sementara itu, parametricism membahas cara pembelajaran disimpan dan sering kali juga metode pembelajarannya.

Parametric model memiliki jumlah parameter tetap, sedangkan nonparametric model memiliki jumlah parameter yang tidak terbatas dan ditentukan oleh data.

Sebagai contoh, bayangkan masalahnya adalah memasukkan pasak berbentuk persegi ke lubang persegi yang tepat. Sebagian manusia, seperti bayi, mungkin mencoba memasukkannya ke semua lubang sampai cocok. Ini menyerupai pendekatan parametric. Seorang remaja mungkin menghitung jumlah sisi, yaitu empat, lalu mencari lubang dengan jumlah sisi yang sama. Ini menyerupai pendekatan nonparametric.

Parametric models cenderung menggunakan trial and error, sedangkan nonparametric models cenderung melakukan counting.

## Supervised parametric learning

Secara sederhana: trial-and-error learning menggunakan knobs.

Supervised parametric learning machines adalah mesin dengan jumlah knobs tetap. Bagian fixed knobs inilah yang membuatnya parametric. Proses belajar terjadi dengan memutar knobs tersebut.

Data input masuk, diproses berdasarkan sudut knobs, lalu diubah menjadi sebuah prediction.

Learning dilakukan dengan memutar knobs ke sudut yang berbeda. Misalnya, model mencoba memprediksi kemungkinan Red Sox memenangkan World Series. Model pertama-tama mengambil data, seperti statistik olahraga berupa catatan menang/kalah atau rata-rata jumlah jari kaki per pemain, lalu menghasilkan prediksi seperti peluang 98%.

Setelah itu, model mengamati apakah Red Sox benar-benar menang atau tidak. Setelah mengetahui hasilnya, learning algorithm akan memperbarui knobs agar prediksi berikutnya menjadi lebih akurat ketika melihat data input yang sama atau mirip.

Jika catatan menang/kalah adalah predictor yang baik, model mungkin menaikkan knob tersebut. Sebaliknya, jika rata-rata jumlah jari kaki bukan predictor yang baik, model mungkin menurunkan knob tersebut. Beginilah parametric models belajar.

Seluruh hal yang dipelajari model dapat ditangkap dari posisi knobs pada waktu tertentu. Model seperti ini juga dapat dipahami sebagai search algorithm, karena prosesnya mencari konfigurasi knobs yang tepat dengan mencoba, menyesuaikan, lalu mencoba lagi.

Trial and error bukan definisi formal, tetapi merupakan sifat umum pada parametric models. Ketika ada jumlah knobs yang tetap, dibutuhkan proses pencarian untuk menemukan konfigurasi optimal. Ini berbeda dengan nonparametric learning, yang sering berbasis counting dan kurang lebih menambahkan knobs baru ketika menemukan sesuatu yang baru untuk dihitung.

## Step 1: Predict

Untuk menjelaskan supervised parametric learning, contoh olahraga digunakan lagi. Model mencoba memprediksi apakah Red Sox akan memenangkan World Series.

Langkah pertama adalah mengumpulkan statistik olahraga, memasukkannya ke mesin, dan membuat prediksi tentang probabilitas Red Sox menang.

## Step 2: Compare to the truth pattern

Langkah kedua adalah membandingkan prediksi dengan pola kebenaran yang diperhatikan, yaitu apakah Red Sox menang.

Dalam contoh ini, prediksinya adalah 98%, tetapi Red Sox kalah. Maka perbandingannya adalah:

**Pred: 98% > Truth: 0%**

Langkah ini menunjukkan bahwa jika model memprediksi 0%, maka model akan memprediksi kekalahan tersebut secara sempurna. Tujuannya adalah membuat mesin menjadi akurat.

## Step 3: Learn the pattern

Langkah ini menyesuaikan knobs dengan mempelajari seberapa jauh model meleset dan seperti apa data input pada saat prediksi dibuat.

Setelah itu, knobs diputar agar prediksi menjadi lebih akurat untuk data input tersebut.

Secara teori, ketika model melihat statistik olahraga yang sama lagi, prediksinya akan lebih rendah dari 98%. Setiap knob merepresentasikan sensitivitas prediksi terhadap jenis data input yang berbeda. Hal itulah yang diubah ketika model “belajar”.

## Unsupervised parametric learning

Unsupervised parametric learning menggunakan pendekatan yang sangat mirip. Pada tingkat tinggi, unsupervised learning berfokus pada pengelompokan data. Unsupervised parametric learning menggunakan knobs untuk mengelompokkan data.

Dalam kasus ini, biasanya ada beberapa knobs untuk setiap grup. Setiap knob memetakan kecocokan input data terhadap grup tertentu. Ini adalah deskripsi tingkat tinggi.

Contohnya, anggap data ingin dibagi menjadi tiga grup. Dalam dataset, ada tiga cluster yang ingin ditemukan oleh parametric model. Cluster tersebut ditunjukkan sebagai group 1, group 2, dan group 3.

Ketika datapoint pertama diproses melalui model unsupervised yang sudah dilatih, datapoint tersebut paling kuat dipetakan ke group 1.

Setiap machine milik grup mencoba mengubah input data menjadi angka antara 0 dan 1. Angka tersebut memberi tahu probabilitas bahwa input data merupakan anggota grup tersebut.

Ada banyak variasi tentang bagaimana model seperti ini dilatih dan seperti apa hasil akhirnya. Namun, pada tingkat tinggi, model menyesuaikan parameter untuk mengubah input data menjadi grup yang sesuai.

## Nonparametric learning

Secara sederhana: counting-based methods.

Nonparametric learning adalah kelas algoritma yang jumlah parameternya didasarkan pada data, bukan ditentukan sebelumnya. Karena itu, metode ini umumnya melakukan counting dengan satu cara atau lainnya, sehingga jumlah parameter bertambah berdasarkan jumlah item yang dihitung dalam data.

Dalam supervised setting, misalnya, nonparametric model dapat menghitung berapa kali warna lampu lalu lintas tertentu membuat mobil “jalan”. Setelah menghitung beberapa contoh saja, model dapat memprediksi bahwa lampu tengah selalu membuat mobil jalan, sedangkan lampu kanan hanya kadang-kadang membuat mobil jalan.

Model tersebut memiliki tiga parameter, yaitu tiga hitungan yang menunjukkan berapa kali setiap warna lampu menyala dan mobil berjalan, mungkin dibagi dengan jumlah total observasi.

Jika ada lima lampu, maka akan ada lima hitungan atau lima parameter. Sifat yang membuat model sederhana ini nonparametric adalah jumlah parameternya berubah berdasarkan data. Dalam contoh ini, jumlah parameter berubah berdasarkan jumlah lampu.

Ini berbeda dengan parametric models, yang memulai dengan jumlah parameter tertentu. Jumlah parameter pada parametric models dapat lebih banyak atau lebih sedikit berdasarkan keputusan scientist yang melatih model, bukan berdasarkan data.

Ada area abu-abu antara parametric dan nonparametric algorithms. Parametric models sebelumnya terlihat memiliki knob untuk setiap input datapoint. Sebagian besar parametric models tetap memiliki semacam input berdasarkan jumlah kelas dalam data. Artinya, bahkan parametric algorithms pun dapat dipengaruhi oleh jumlah kelas dalam data, meskipun tidak secara eksplisit menghitung pola.

Hal ini juga menunjukkan bahwa parameters adalah istilah umum untuk kumpulan angka yang digunakan untuk memodelkan pola. Counts adalah parameters. Weights adalah parameters. Variasi counts atau weights yang dinormalisasi juga parameters. Correlation coefficients dapat menjadi parameters.

Deep learning adalah kelas dari parametric models. Buku ini tidak akan membahas nonparametric models lebih lanjut, tetapi kelas algoritma tersebut tetap menarik dan kuat.
