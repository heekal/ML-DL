## 2. Random Sampling dan Sample Bias

**Sample** adalah subset dari data yang lebih besar. Data yang lebih besar disebut **population**. Dalam statistik, population berarti kumpulan data besar yang didefinisikan, bukan populasi biologis.

**Random sampling** adalah proses mengambil elemen dari populasi dengan peluang yang sama untuk dipilih pada setiap pengambilan. Hasilnya disebut **simple random sample**. Sampling dapat dilakukan:

- **With replacement**: observasi dikembalikan ke populasi setelah dipilih, sehingga dapat terpilih lagi.
- **Without replacement**: observasi yang sudah dipilih tidak tersedia untuk pengambilan berikutnya.

Kualitas data sering lebih penting daripada jumlah data. Dalam data science, kualitas data berkaitan dengan kelengkapan, konsistensi format, kebersihan, dan akurasi titik data. Dalam statistik, ada tambahan konsep **representativeness** atau keterwakilan.

### Key Terms: Random Sampling

| Istilah | Makna |
|---|---|
| Sample | Subset dari data yang lebih besar. |
| Population | Data yang lebih besar atau gagasan tentang data yang lebih besar. |
| N / n | Ukuran populasi / sampel. |
| Random sampling | Mengambil elemen ke dalam sampel secara acak. |
| Stratified sampling | Membagi populasi ke dalam strata, lalu mengambil sampel acak dari setiap strata. |
| Stratum / strata | Subkelompok homogen dalam populasi dengan karakteristik yang sama. |
| Simple random sample | Sampel dari random sampling tanpa stratifikasi. |
| Bias | Error sistematis. |
| Sample bias | Sampel yang salah merepresentasikan populasi. |

### Contoh Sample Bias

Contoh klasik adalah polling **Literary Digest** tahun 1936 yang memprediksi Alf Landon menang atas Franklin Roosevelt. Literary Digest melakukan polling ke lebih dari 10 juta orang, tetapi prediksinya salah. George Gallup hanya melakukan polling kepada sekitar 2.000 orang dan memprediksi Roosevelt menang dengan benar.

Masalah utama Literary Digest bukan jumlah data, tetapi cara memilih responden. Sampelnya cenderung berasal dari kelompok sosial ekonomi lebih tinggi, seperti pelanggan majalah, pemilik telepon, dan pemilik mobil. Ini menghasilkan **sample bias**, karena sampel berbeda secara bermakna dan tidak acak dari populasi pemilih yang ingin direpresentasikan.

### Self-Selection Sampling Bias

Review restoran, hotel, atau kafe di media sosial rawan bias karena orang yang menulis review tidak dipilih secara acak. Mereka memilih sendiri untuk menulis. Ini disebut **self-selection bias**. Orang yang termotivasi menulis review bisa saja memiliki pengalaman buruk, memiliki hubungan dengan tempat tersebut, atau berbeda dari orang yang tidak menulis review.

### Bias

**Statistical bias** adalah error pengukuran atau sampling yang bersifat sistematis dan muncul dari proses pengukuran atau proses sampling.

Perlu dibedakan antara error karena peluang acak dan error karena bias. Proses yang tidak bias tetap bisa menghasilkan error, tetapi error tersebut acak dan tidak mengarah kuat ke satu arah tertentu. Proses yang bias menghasilkan error acak sekaligus kecenderungan sistematis ke arah tertentu.

Jika hasil menunjukkan bias, itu dapat menjadi tanda bahwa model statistik atau machine learning salah dispesifikasikan, atau ada variabel penting yang tertinggal.

### Random Selection

Random sampling membutuhkan definisi populasi yang jelas dan prosedur sampling yang jelas. Misalnya, ketika ingin membuat profil pelanggan, perlu ditentukan siapa yang dianggap pelanggan: semua yang pernah membeli, yang pembeliannya lebih dari 0, termasuk refund atau tidak, termasuk pembelian internal atau tidak, dan sebagainya.

Jika sampling dilakukan dari data yang mengalir, seperti transaksi real-time atau pengunjung web, waktu juga penting. Pengunjung pada pukul 10 pagi hari kerja bisa berbeda dari pengunjung pukul 10 malam akhir pekan.

Pada **stratified sampling**, populasi dibagi menjadi beberapa strata dan sampel acak diambil dari setiap strata. Contohnya, polling politik dapat memberi bobot lebih besar pada strata yang jumlahnya kecil agar ukuran sampelnya cukup untuk dianalisis.

### Size Versus Quality

Dalam era big data, data yang lebih kecil kadang lebih baik. Random sampling dapat mengurangi bias dan memungkinkan eksplorasi serta peningkatan kualitas data yang lebih teliti. Missing value dan outlier mungkin mengandung informasi penting, tetapi memeriksa semuanya pada jutaan record bisa terlalu mahal. Pada sampel beberapa ribu record, pemeriksaan manual masih mungkin dilakukan.

Data sangat besar dibutuhkan ketika data tidak hanya besar, tetapi juga sparse. Karena kombinasi query tertentu bisa sangat jarang muncul, jumlah data sangat besar dibutuhkan agar hasil pencarian tetap efektif.

### Sample Mean Versus Population Mean

Simbol **x-bar** digunakan untuk mean sampel, sedangkan **μ** digunakan untuk mean populasi. Perbedaan ini penting karena informasi tentang sampel dapat diamati langsung, sedangkan informasi tentang populasi besar sering disimpulkan dari sampel kecil.

### Key Ideas: Random Sampling dan Sample Bias

## 3. Selection Bias

**Selection bias** adalah praktik memilih data secara selektif, sadar atau tidak sadar, sehingga menghasilkan kesimpulan yang menyesatkan atau sementara.

### Key Terms: Selection Bias

| Istilah | Makna |
|---|---|
| Selection bias | Bias yang muncul dari cara observasi dipilih. |
| Data snooping | Pencarian ekstensif dalam data untuk menemukan sesuatu yang menarik. |
| Vast search effect | Bias atau hasil yang tidak dapat direproduksi akibat pemodelan berulang atau pemodelan dengan banyak variabel prediktor. |

Jika hipotesis ditentukan lebih dulu dan eksperimen dirancang baik untuk mengujinya, kesimpulan bisa lebih dipercaya. Masalah sering muncul ketika data yang tersedia diperiksa berulang-ulang sampai ditemukan pola yang terlihat menarik. Pola itu bisa nyata, tetapi bisa juga hanya hasil keberuntungan.

Materi memberi contoh pelemparan koin: seseorang yang mengklaim bisa mendapatkan 10 kepala berturut-turut lalu berhasil melakukannya akan terlihat istimewa. Tetapi jika 20.000 orang di stadion masing-masing melempar koin 10 kali, sangat mungkin ada seseorang yang mendapatkan 10 kepala berturut-turut hanya karena peluang.

Dalam data science, selection bias perlu diwaspadai karena analisis terhadap dataset besar sering dilakukan berulang. Cara menguranginya adalah memakai **holdout set** untuk validasi, dan dapat juga menggunakan **target shuffling** atau permutation test untuk menguji validitas asosiasi prediktif.

### Regression to the Mean

**Regression to the mean** adalah fenomena ketika observasi ekstrem cenderung diikuti oleh observasi yang lebih mendekati pusat. Memberi makna berlebihan pada nilai ekstrem dapat menyebabkan selection bias.

Contoh dari materi adalah fenomena “rookie of the year, sophomore slump”. Atlet rookie terbaik biasanya memiliki kombinasi skill dan keberuntungan. Pada musim berikutnya, skill mungkin tetap ada, tetapi keberuntungan tidak selalu terulang, sehingga performanya turun menuju nilai yang lebih biasa.

Regression to the mean berbeda dari metode statistik linear regression. Dalam konteks ini, regression berarti “kembali” ke nilai yang lebih pusat.

### Key Ideas: Selection Bias

- Menentukan hipotesis sebelum mengumpulkan data dengan prinsip randomization dan random sampling membantu mencegah bias.
- Analisis data yang dilakukan setelah melihat data, termasuk data snooping dan pemilihan kejadian menarik setelah fakta terjadi, memiliki risiko bias.

## 4. Sampling Distribution of a Statistic

**Sampling distribution of a statistic** adalah distribusi dari suatu statistik sampel ketika banyak sampel diambil dari populasi yang sama. Statistik klasik banyak membahas cara menyimpulkan informasi dari sampel kecil ke populasi yang jauh lebih besar.

### Key Terms: Sampling Distribution

| Istilah | Makna |
|---|---|
| Sample statistic | Metrik yang dihitung dari sampel data. |
| Data distribution | Distribusi frekuensi nilai individual dalam dataset. |
| Sampling distribution | Distribusi frekuensi statistik sampel dari banyak sampel atau resample. |
| Central limit theorem | Kecenderungan sampling distribution menjadi berbentuk normal ketika ukuran sampel meningkat. |
| Standard error | Variabilitas statistik sampel pada banyak sampel. |

Sampel biasanya diambil untuk mengukur sesuatu dengan statistik sampel atau membangun model. Karena estimasi atau model dibuat dari sampel, hasilnya bisa berbeda jika sampel lain diambil. Hal ini disebut **sampling variability**.

Distribusi titik data individual disebut **data distribution**, sedangkan distribusi statistik sampel disebut **sampling distribution**. Sampling distribution dari mean biasanya lebih teratur dan lebih berbentuk lonceng dibandingkan distribusi datanya. Semakin besar ukuran sampel, distribusi statistik sampel makin sempit.

### Kode: Sampling Distribution dari Income

Kode pada materi menggunakan `loans_income`. PDF menjelaskan data ini sebagai annual income dari loan applicants LendingClub. Dataset atau variabel `loans_income` tidak disertakan pada PDF, jadi cell ini dijalankan setelah variabel tersebut tersedia.

```python
if 'loans_income' not in globals():
  print('Variabel loans_income belum tersedia. Kode ini mengikuti materi dan membutuhkan data loans_income.')
else:
  sample_data = pd.DataFrame({
    'income': loans_income.sample(1000),
    'type': 'Data',
  })

  sample_mean_05 = pd.DataFrame({
    'income': [loans_income.sample(5).mean() for _ in range(1000)],
    'type': 'Mean of 5',
  })

  sample_mean_20 = pd.DataFrame({
    'income': [loans_income.sample(20).mean() for _ in range(1000)],
    'type': 'Mean of 20',
  })

  results = pd.concat([sample_data, sample_mean_05, sample_mean_20])

  g = sns.FacetGrid(results, col='type', col_wrap=1, height=2, aspect=2)
  g.map(plt.hist, 'income', range=[0, 200000], bins=40)
  g.set_axis_labels('Income', 'Count')
  g.set_titles('{col_name}')
```

### Central Limit Theorem

**Central limit theorem** menyatakan bahwa mean dari banyak sampel akan menyerupai kurva normal berbentuk lonceng, walaupun populasi asalnya tidak normal, selama ukuran sampel cukup besar dan penyimpangan data dari normalitas tidak terlalu besar.

Teorema ini memungkinkan penggunaan formula pendekatan normal, seperti t-distribution, dalam confidence interval dan hypothesis test. Dalam praktik data science, central limit theorem tetap penting untuk dipahami, tetapi bootstrap sering menjadi cara praktis untuk menjawab pertanyaan tentang sampling error.

### Standard Error

**Standard error** merangkum variabilitas sampling distribution dari suatu statistik.

Rumus standard error:

$$
SE = \frac{s}{\sqrt{n}}
$$

Dengan bertambahnya ukuran sampel, standard error menurun. Untuk menurunkan standard error sebesar faktor 2, ukuran sampel harus dinaikkan sebesar faktor 4. Ini disebut **square root of n rule**.

Standard deviation mengukur variabilitas titik data individual. Standard error mengukur variabilitas metrik sampel.

### Key Ideas: Sampling Distribution

- Distribusi frekuensi statistik sampel menunjukkan bagaimana metrik dapat berubah dari sampel ke sampel.
- Sampling distribution dapat diestimasi dengan bootstrap atau formula berbasis central limit theorem.
- Standard error adalah metrik penting untuk merangkum variabilitas statistik sampel.

## 5. The Bootstrap

**Bootstrap** adalah cara mengestimasi sampling distribution dari suatu statistik atau parameter model dengan mengambil sampel tambahan dari sampel yang sudah ada, dengan replacement, lalu menghitung ulang statistik atau model pada setiap resample.

Bootstrap tidak harus mengasumsikan data atau statistik sampel berdistribusi normal.

### Key Terms: Bootstrap

| Istilah | Makna |
|---|---|
| Bootstrap sample | Sampel yang diambil dengan replacement dari dataset yang diamati. |
| Resampling | Proses mengambil sampel berulang dari data yang diamati, mencakup bootstrap dan permutation/shuffling. |

### Algoritma Bootstrap untuk Mean

Untuk sampel berukuran `n`:

1. Ambil satu nilai sampel, catat, lalu kembalikan.
2. Ulangi sebanyak `n` kali.
3. Catat mean dari `n` nilai hasil resampling.
4. Ulangi langkah 1-3 sebanyak `R` kali.
5. Gunakan hasil `R` kali untuk menghitung standard deviation, membuat histogram/boxplot, atau mencari confidence interval.

Bootstrap tidak menciptakan data baru dan tidak memperbaiki sampel kecil. Bootstrap hanya memberi gambaran bagaimana banyak sampel tambahan akan berperilaku jika diambil dari populasi yang mirip dengan sampel asli.

### Kode: Bootstrap Median

Kode pada materi memakai `resample` dari scikit-learn dan variabel `loans_income`. Jalankan setelah `loans_income` tersedia.

```python
if 'loans_income' not in globals():
  print('Variabel loans_income belum tersedia. Kode ini mengikuti materi dan membutuhkan data loans_income.')
else:
  results = []
  for nrepeat in range(1000):
    sample = resample(loans_income)
    results.append(sample.median())

  results = pd.Series(results)

  print('Bootstrap Statistics:')
  print(f'original: {loans_income.median()}')
  print(f'bias: {results.mean() - loans_income.median()}')
  print(f'std. error: {results.std()}')
```

### Resampling Versus Bootstrapping

Resampling kadang digunakan sebagai sinonim bootstrap, tetapi sering juga mencakup prosedur permutation, yaitu sampel digabung dan pengambilan dapat dilakukan tanpa replacement. Bootstrap selalu berarti sampling dengan replacement dari data yang diamati.

### Key Ideas: Bootstrap

- Bootstrap adalah alat yang kuat untuk menilai variabilitas statistik sampel.
- Bootstrap dapat diterapkan pada banyak situasi tanpa membutuhkan pendekatan matematis khusus terhadap sampling distribution.
- Bootstrap dapat mengestimasi sampling distribution untuk statistik yang belum memiliki pendekatan matematis.
- Pada model prediktif, penggabungan prediksi dari banyak bootstrap sample disebut bagging dan sering lebih baik daripada satu model tunggal.

## 6. Confidence Intervals

**Confidence interval** adalah cara menyajikan estimasi sebagai rentang, bukan satu angka tunggal. Ini membantu menunjukkan ketidakpastian dalam estimasi sampel.

### Key Terms: Confidence Intervals

| Istilah | Makna |
|---|---|
| Confidence level | Persentase confidence interval yang dibuat dengan cara sama dari populasi yang sama dan diharapkan memuat statistik yang menjadi perhatian. |
| Interval endpoints | Batas bawah dan batas atas confidence interval. |

Confidence interval selalu memiliki coverage level, misalnya 90% atau 95%. Salah satu cara memahami 90% confidence interval adalah: interval tersebut memuat 90% bagian tengah dari bootstrap sampling distribution suatu statistik sampel.

Algoritma bootstrap confidence interval:

1. Ambil random sample berukuran `n` dengan replacement dari data.
2. Catat statistik yang diinginkan dari resample.
3. Ulangi langkah 1-2 sebanyak `R` kali.
4. Untuk confidence interval `x%`, buang `[(100-x)/2]%` hasil resample dari setiap sisi distribusi.
5. Titik pemotongan tersebut menjadi endpoint confidence interval.

Semakin tinggi confidence level, interval semakin lebar. Semakin kecil sampel, interval juga semakin lebar.

### Key Ideas: Confidence Intervals

- Confidence interval adalah cara umum menyajikan estimasi sebagai rentang.
- Semakin banyak data, semakin kecil variabilitas estimasi sampel.
- Semakin rendah confidence level yang dapat ditoleransi, semakin sempit confidence interval.
- Bootstrap adalah cara efektif untuk membangun confidence interval.

## 7. Normal Distribution

**Normal distribution** adalah distribusi berbentuk lonceng yang sangat penting dalam statistik tradisional. Banyak distribusi statistik sampel sering berbentuk normal, sehingga normal distribution berguna dalam formula matematis untuk memperkirakan uncertainty dan variability.

### Key Terms: Normal Distribution

| Istilah | Makna |
|---|---|
| Error | Selisih antara titik data dan nilai prediksi atau rata-rata. |
| Standardize | Mengurangi mean dan membagi dengan standard deviation. |
| z-score | Hasil standardisasi titik data. |
| Standard normal | Normal distribution dengan mean = 0 dan standard deviation = 1. |
| QQ-Plot | Plot untuk melihat seberapa dekat distribusi sampel dengan distribusi tertentu, misalnya normal distribution. |

Pada normal distribution, sekitar 68% data berada dalam satu standard deviation dari mean, dan sekitar 95% berada dalam dua standard deviation dari mean.

Normal distribution juga disebut Gaussian distribution. Dalam statistik, error adalah selisih antara nilai aktual dan estimasi statistik seperti mean sampel.

Miskonsepsi penting: normal distribution bukan disebut normal karena sebagian besar data nyata mengikutinya. Banyak data mentah dalam data science tidak berdistribusi normal. Kegunaan normal distribution terutama karena banyak statistik memiliki sampling distribution yang normal.

### Standard Normal dan QQ-Plot

Standard normal distribution memakai satuan standard deviation dari mean pada sumbu x. Untuk membandingkan data dengan standard normal distribution, kurangi nilai dengan mean lalu bagi dengan standard deviation. Hasilnya disebut **z-score**.

QQ-Plot digunakan untuk melihat seberapa dekat suatu sampel dengan distribusi tertentu. Jika titik-titik mengikuti garis diagonal, sampel dapat dianggap mendekati distribusi tersebut.

### Kode: QQ-Plot dari Sampel Normal

```python
fig, ax = plt.subplots(figsize=(4, 4))
norm_sample = stats.norm.rvs(size=100)
stats.probplot(norm_sample, plot=ax)
plt.show()
```

Mengubah data menjadi z-score tidak membuat data menjadi normal. Standardisasi hanya membuat data berada pada skala yang sama dengan standard normal distribution untuk tujuan perbandingan.

### Key Ideas: Normal Distribution

- Normal distribution penting dalam sejarah statistik karena memungkinkan pendekatan matematis terhadap uncertainty dan variability.
- Raw data biasanya tidak berdistribusi normal, tetapi error, average, dan total dalam sampel besar sering normal.
- z-score diperoleh dengan mengurangi mean dan membagi dengan standard deviation.

## 9. Student's t-Distribution

**t-distribution** berbentuk seperti normal distribution, tetapi memiliki tail yang sedikit lebih tebal dan panjang. Distribusi ini banyak digunakan untuk menggambarkan distribusi statistik sampel.

Distribusi mean sampel biasanya berbentuk seperti t-distribution. Ada keluarga t-distribution yang berbeda menurut ukuran sampel. Semakin besar sampel, bentuk t-distribution semakin mendekati normal distribution.

### Key Terms: Student's t-Distribution

| Istilah | Makna |
|---|---|
| n | Ukuran sampel. |
| Degrees of freedom | Parameter yang memungkinkan t-distribution menyesuaikan ukuran sampel, statistik, dan jumlah grup. |

t-distribution disebut Student's t karena dipublikasikan oleh W. S. Gosset dengan nama “Student”. Distribusi ini digunakan untuk memperkirakan confidence interval dengan mempertimbangkan sampling variation.

Untuk sampel berukuran `n` dengan mean sampel `x-bar` dan sample standard deviation `s`, 90% confidence interval di sekitar mean sampel dituliskan sebagai:

$$
\bar{x} \pm t_{n-1, 0.05} \cdot \frac{s}{\sqrt{n}}
$$

Bagi data scientist, t-distribution tidak selalu menjadi pusat praktik karena bootstrap dapat menjawab banyak pertanyaan sampling error secara empiris. Namun, t-statistic sering muncul dalam output software statistik, misalnya A/B test dan regression.

### Key Ideas: Student's t-Distribution

- t-distribution adalah keluarga distribusi yang mirip normal distribution tetapi dengan tail lebih tebal.
- t-distribution digunakan sebagai referensi untuk distribusi mean sampel, perbedaan dua mean sampel, parameter regression, dan statistik lainnya.

## 10. Binomial Distribution

Outcome yes/no atau 0/1 penting dalam analytics karena sering menjadi hasil akhir dari keputusan atau proses, seperti buy/don't buy, click/don't click, survive/die, dan sebagainya.

Binomial distribution didasarkan pada sekumpulan trial, dan setiap trial memiliki dua outcome dengan probabilitas tertentu. Contoh: melempar koin 10 kali adalah binomial experiment dengan 10 trial dan dua outcome pada setiap trial.

### Key Terms: Binomial Distribution

| Istilah | Makna |
|---|---|
| Trial | Kejadian dengan outcome diskrit, misalnya coin flip. |
| Success | Outcome yang menjadi perhatian; sinonimnya “1”. |
| Binomial | Memiliki dua outcome; sinonimnya yes/no, 0/1, binary. |
| Binomial trial | Trial dengan dua outcome; sinonimnya Bernoulli trial. |
| Binomial distribution | Distribusi jumlah success dalam `x` trial; sinonimnya Bernoulli distribution. |

Binomial distribution adalah frequency distribution dari jumlah success `x` dalam sejumlah trial `n`, dengan probabilitas success `p` pada setiap trial.

Contoh pertanyaan binomial dari materi: jika probabilitas sebuah click berubah menjadi sale adalah 0.02, berapa probabilitas mengamati 0 sale dalam 200 click?

Mean binomial distribution adalah:

$$
n 	* p
$$

Variance binomial distribution adalah:

$$
n 	* p(1-p)
$$

Dengan jumlah trial cukup besar, terutama ketika `p` dekat dengan 0.50, binomial distribution hampir tidak dapat dibedakan dari normal distribution.

### Kode: Binomial Probability

```python
stats.binom.pmf(2, n=5, p=0.1)
```

Kode di atas menghitung probabilitas mengamati tepat `x = 2` success dalam `n = 5` trial, dengan probabilitas success `p = 0.1`.

```python
stats.binom.cdf(2, n=5, p=0.1)
```

Kode di atas menghitung probabilitas mengamati `2` atau lebih sedikit success dalam `5` trial, dengan probabilitas success `0.1`.

### Key Ideas: Binomial Distribution

- Outcome binomial penting karena merepresentasikan keputusan dasar seperti buy/don't buy, click/don't click, survive/die.
- Binomial trial adalah eksperimen dengan dua outcome: satu dengan probabilitas `p`, dan satu lagi dengan probabilitas `1-p`.
- Dengan `n` besar dan `p` tidak terlalu dekat dengan 0 atau 1, binomial distribution dapat didekati oleh normal distribution.

## 11. Chi-Square Distribution

Chi-square distribution berkaitan dengan penyimpangan dari ekspektasi, terutama pada hitungan kategori.

Ekspektasi secara longgar berarti tidak ada hal yang tidak biasa pada data, misalnya tidak ada korelasi antara variabel atau tidak ada pola prediktif. Ini juga disebut **null hypothesis** atau **null model**.

Chi-square statistic mengukur seberapa jauh nilai yang diamati menyimpang dari nilai yang diharapkan. Statistik ini berguna untuk menentukan apakah beberapa treatment, seperti A/B/C test, berbeda satu sama lain dalam efeknya.

### Key Ideas: Chi-Square Distribution

- Chi-square distribution biasanya berkaitan dengan hitungan subjek atau item yang masuk ke kategori.
- Chi-square statistic mengukur sejauh mana data menyimpang dari apa yang diharapkan pada null model.

## 12. F-Distribution

F-distribution digunakan dalam eksperimen ilmiah ketika beberapa treatment diuji pada beberapa grup, misalnya pupuk berbeda pada blok lahan berbeda. Ini mirip A/B/C test, tetapi datanya berupa nilai kontinu, bukan hitungan.

F-statistic mengukur apakah perbedaan antar mean grup lebih besar daripada yang diharapkan dari variasi acak normal. F-statistic adalah rasio antara variabilitas antar mean grup dan variabilitas di dalam masing-masing grup. Perbandingan ini disebut **analysis of variance** atau ANOVA.

F-statistic juga digunakan dalam linear regression untuk membandingkan variasi yang dijelaskan oleh model dengan variasi keseluruhan pada data.

### Key Ideas: F-Distribution

- F-distribution digunakan pada eksperimen dan linear model yang melibatkan data terukur.
- F-statistic membandingkan variasi karena faktor yang menjadi perhatian dengan variasi keseluruhan.

## 13. Poisson and Related Distributions

Banyak proses menghasilkan event secara acak pada rate tertentu, misalnya pengunjung website, mobil yang tiba di gerbang tol, cacat pada kain, atau typo per 100 baris kode.

### Key Terms: Poisson and Related Distributions

| Istilah | Makna |
|---|---|
| Lambda | Rate event per unit waktu atau ruang. |
| Poisson distribution | Distribusi frekuensi jumlah event dalam unit waktu atau ruang. |
| Exponential distribution | Distribusi frekuensi waktu atau jarak dari satu event ke event berikutnya. |
| Weibull distribution | Generalisasi exponential distribution ketika event rate dapat berubah seiring waktu. |

### Poisson Distribution

Poisson distribution menunjukkan distribusi jumlah event per unit waktu atau ruang ketika banyak unit tersebut disampel. Parameter utama Poisson distribution adalah **λ** atau lambda, yaitu rata-rata jumlah event dalam interval waktu atau ruang tertentu. Variance pada Poisson distribution juga bernilai λ.

Poisson distribution berguna untuk pertanyaan queueing seperti kapasitas server yang dibutuhkan untuk memproses traffic dalam interval lima detik.

### Kode: Random Number dari Poisson Distribution

```python
stats.poisson.rvs(2, size=100)
```

Kode di atas menghasilkan 100 random number dari Poisson distribution dengan λ = 2. Jika incoming customer service calls rata-rata dua per menit, kode ini mensimulasikan 100 menit dan mengembalikan jumlah call pada setiap menit.

### Exponential Distribution

Dengan parameter λ yang sama seperti Poisson distribution, exponential distribution dapat memodelkan waktu antara event, seperti waktu antar kunjungan website atau waktu antar mobil tiba di gerbang tol. Distribusi ini juga digunakan untuk memodelkan time to failure dan durasi service call.

### Kode: Random Number dari Exponential Distribution

```python
stats.expon.rvs(0.2, size=100)
```

Kode di atas menghasilkan 100 random number dari exponential distribution ketika rata-rata jumlah event per periode waktu adalah 0.2.

Asumsi penting pada Poisson dan exponential distribution adalah rate λ tetap konstan selama periode yang dianalisis. Dalam praktik, rate sering berubah menurut waktu atau ruang, sehingga analisis dapat dibagi menjadi segmen yang lebih homogen.

### Estimating the Failure Rate

Dalam banyak aplikasi, event rate λ diketahui atau dapat diestimasi dari data sebelumnya. Untuk event langka, seperti kegagalan mesin pesawat, data mungkin terlalu sedikit untuk mengestimasi rate secara presisi. Jika ada sedikit data tetapi belum cukup andal, goodness-of-fit test dapat digunakan untuk melihat seberapa baik beberapa rate cocok dengan data yang diamati.

### Weibull Distribution

Jika event rate berubah selama interval, exponential atau Poisson distribution tidak lagi cocok. Contohnya pada kegagalan mekanis, risiko failure meningkat seiring waktu.

Weibull distribution adalah perluasan dari exponential distribution dengan event rate yang boleh berubah, ditentukan oleh parameter bentuk **β**. Jika β > 1, probabilitas event meningkat seiring waktu. Jika β < 1, probabilitas event menurun.

Parameter kedua adalah **η** atau eta, disebut juga scale parameter atau characteristic life. Pada Weibull, proses estimasi mencakup β dan η.

### Kode: Random Number dari Weibull Distribution

```python
stats.weibull_min.rvs(1.5, scale=5000, size=100)
```

Kode di atas menghasilkan 100 random number atau lifetime dari Weibull distribution dengan shape 1.5 dan characteristic life 5000.

### Key Ideas: Poisson and Related Distributions

- Untuk event yang terjadi pada rate konstan, jumlah event per unit waktu atau ruang dapat dimodelkan dengan Poisson distribution.
- Waktu atau jarak antara satu event dan event berikutnya dapat dimodelkan dengan exponential distribution.
- Event rate yang berubah seiring waktu, seperti peningkatan probabilitas kegagalan perangkat, dapat dimodelkan dengan Weibull distribution.

## 14. Summary

Prinsip random sampling tetap penting dalam era big data ketika estimasi yang akurat dibutuhkan. Random selection dapat mengurangi bias dan menghasilkan data yang lebih berkualitas daripada hanya memakai data yang tersedia secara nyaman.

Pengetahuan tentang sampling distribution dan data-generating distribution membantu mengukur kemungkinan error dalam estimasi akibat variasi acak. Bootstrap, yaitu sampling dengan replacement dari data yang diamati, menjadi metode umum yang fleksibel untuk memperkirakan kemungkinan error pada estimasi sampel.
