## 1. Regression and Prediction

Tujuan umum statistik adalah menjawab apakah variabel **X** atau beberapa variabel predictor berhubungan dengan variabel **Y**, bagaimana bentuk hubungannya, dan apakah hubungan itu dapat digunakan untuk memprediksi **Y**.

Dalam data science, regression sangat dekat dengan **prediction**. Model dilatih pada data yang outcome-nya sudah diketahui, lalu digunakan untuk data baru yang outcome-nya belum diketahui. Proses ini disebut **supervised learning**. Regression diagnostics juga dapat digunakan untuk mendeteksi record yang tidak biasa atau anomali.

## 2. Simple Linear Regression

**Simple linear regression** memodelkan hubungan antara satu predictor dan satu response. Jika **X** naik, **Y** dapat naik atau turun. Correlation mengukur kekuatan asosiasi dua variabel, sedangkan regression mengukur bentuk hubungannya.

| Istilah | Makna |
|---|---|
| Response | Variabel yang ingin diprediksi; disebut juga dependent variable, Y variable, target, atau outcome. |
| Independent variable | Variabel yang digunakan untuk memprediksi response; disebut juga X variable, feature, attribute, atau predictor. |
| Record | Vektor nilai predictor dan outcome untuk satu individu atau kasus. |
| Intercept | Nilai prediksi saat X = 0; disebut juga b0 atau beta0. |
| Regression coefficient | Slope garis regression; disebut juga b1, beta1, parameter estimates, atau weights. |
| Fitted values | Nilai prediksi dari garis regression. |
| Residuals | Selisih antara nilai aktual dan fitted values. |
| Least squares | Metode fitting regression dengan meminimalkan jumlah residual kuadrat. |

### Regression Equation

Simple linear regression memperkirakan perubahan **Y** saat **X** berubah:

$$Y = b_0 + b_1X$$

**b0** adalah intercept, yaitu nilai prediksi ketika X = 0. **b1** adalah slope untuk X. Variabel **Y** disebut response atau dependent variable, sedangkan **X** disebut predictor atau independent variable.

Contoh dari materi menggunakan data paparan debu kapas: **Exposure** sebagai predictor dan **PEFR** sebagai ukuran kapasitas paru sebagai response.

$$PEFR = b_0 + b_1Exposure$$

```python
if 'lung' in globals():
  predictors = ['Exposure']
  outcome = 'PEFR'

  model = LinearRegression()
  model.fit(lung[predictors], lung[outcome])

  print(f'Intercept: {model.intercept_:.3f}')
  print(f'Coefficient Exposure: {model.coef_[0]:.3f}')
else:
  print('Data lung tidak dicantumkan lengkap di PDF.')
```

Output pada materi menunjukkan intercept sebesar **424.583** dan coefficient Exposure sebesar **-4.185**. Artinya, untuk setiap tambahan satu tahun paparan debu kapas, nilai PEFR diperkirakan turun sebesar 4.185.

### Fitted Values dan Residuals

Karena data biasanya tidak tepat berada pada satu garis, persamaan regression memasukkan error term:

$$Y_i = b_0 + b_1X_i + e_i$$

Fitted values atau predicted values ditulis sebagai:

$$\hat{Y_i} = \hat{b_0} + \hat{b_1}X_i$$

Residual dihitung dari nilai aktual dikurangi nilai prediksi:

$$e_i = Y_i - \hat{Y_i}$$

Notasi topi atau **hat** digunakan untuk membedakan estimasi dari nilai yang diketahui. Estimasi memiliki uncertainty, sedangkan true value dianggap tetap.

```python
if 'lung' in globals() and 'model' in globals():
  fitted = model.predict(lung[predictors])
  residuals = lung[outcome] - fitted
else:
  print('Data lung tidak dicantumkan lengkap di PDF.')
```

### Least Squares

Model regression dipasang dengan mencari garis yang meminimalkan **residual sum of squares** atau **RSS**:

$$RSS = \sum_{i=1}^{n}(Y_i - \hat{Y_i})^2$$

atau:

$$RSS = \sum_{i=1}^{n}(Y_i - \hat{b_0} - \hat{b_1}X_i)^2$$

Metode ini disebut **least squares regression** atau **ordinary least squares (OLS)**. Metode ini cepat dihitung dan banyak digunakan, tetapi sensitif terhadap outlier, terutama pada data kecil atau sedang.

### Prediction Versus Explanation

Secara historis, regression banyak digunakan untuk menjelaskan hubungan linear antara predictor dan outcome. Fokusnya ada pada slope atau coefficient.

Dalam big data dan data science, regression lebih sering digunakan untuk memprediksi outcome individual pada data baru. Fokusnya ada pada fitted values atau nilai prediksi.

Regression dapat menunjukkan hubungan antara X dan Y, tetapi persamaan regression saja tidak membuktikan arah sebab-akibat. Kesimpulan kausalitas harus berasal dari pemahaman yang lebih luas terhadap konteks hubungan tersebut.

**Key Ideas:**

- Regression equation memodelkan hubungan antara response variable Y dan predictor variable X sebagai garis.
- Regression model menghasilkan fitted values dan residuals.
- Regression model biasanya dipasang dengan least squares.
- Regression digunakan untuk prediction dan explanation.

## 3. Multiple Linear Regression

Jika predictor lebih dari satu, persamaan regression diperluas menjadi:

$$Y = b_0 + b_1X_1 + b_2X_2 + ... + b_pX_p + e$$

Model ini tetap linear karena hubungan antara setiap coefficient dan variabelnya bersifat linear.

| Istilah | Makna |
|---|---|
| RMSE | Akar dari rata-rata squared error; metric utama untuk membandingkan regression model. |
| RSE | Mirip RMSE, tetapi disesuaikan dengan degrees of freedom. |
| R-squared | Proporsi variasi data yang dijelaskan oleh model, dari 0 sampai 1. |
| t-statistic | Coefficient predictor dibagi standard error coefficient. |
| Weighted regression | Regression dengan bobot berbeda untuk setiap record. |

### Example: King County Housing Data

Contoh materi menggunakan data rumah King County, Washington, untuk memprediksi **AdjSalePrice** berdasarkan variabel seperti **SqFtTotLiving**, **SqFtLot**, **Bathrooms**, **Bedrooms**, dan **BldgGrade**.

```python
if 'house' in globals():
  subset = ['AdjSalePrice', 'SqFtTotLiving', 'SqFtLot', 'Bathrooms',
            'Bedrooms', 'BldgGrade']
  house[subset].head()
else:
  pd.DataFrame({
    'AdjSalePrice': [300805, 1076162, 761805, 442065, 297065, 411781],
    'SqFtTotLiving': [2400, 3764, 2060, 3200, 1720, 930],
    'SqFtLot': [9373, 20156, 26036, 8618, 8620, 1012],
    'Bathrooms': [3.00, 3.75, 1.75, 3.75, 1.75, 1.50],
    'Bedrooms': [6, 4, 4, 5, 4, 2],
    'BldgGrade': [7, 10, 8, 7, 7, 8]
  })
```

```python
if 'house' in globals():
  predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms', 'BldgGrade']
  outcome = 'AdjSalePrice'

  house_lm = LinearRegression()
  house_lm.fit(house[predictors], house[outcome])

  print(f'Intercept: {house_lm.intercept_:.3f}')
  print('Coefficients:')
  for name, coef in zip(predictors, house_lm.coef_):
    print(f' {name}: {coef}')
else:
  print('Data house tidak dicantumkan lengkap di PDF.')
```

Interpretasi coefficient sama seperti simple linear regression: prediksi **Y** berubah sebesar coefficient **bj** untuk setiap perubahan satu unit pada **Xj**, dengan variabel lain tetap. Contoh dari materi: tambahan satu finished square foot menaikkan estimasi nilai rumah sekitar **$229**.

### Assessing the Model

Metric terpenting dari sudut pandang data science adalah **RMSE**:

$$RMSE = \sqrt{\frac{\sum_{i=1}^{n}(y_i - \hat{y_i})^2}{n}}$$

**RSE** mirip dengan RMSE, tetapi denominator-nya menggunakan degrees of freedom:

$$RSE = \sqrt{\frac{\sum_{i=1}^{n}(y_i - \hat{y_i})^2}{n - p - 1}}$$

Perbedaannya kecil pada data besar.

```python
if 'house' in globals() and 'house_lm' in globals():
  fitted = house_lm.predict(house[predictors])
  RMSE = np.sqrt(mean_squared_error(house[outcome], fitted))
  r2 = r2_score(house[outcome], fitted)

  print(f'RMSE: {RMSE:.0f}')
  print(f'r2: {r2:.4f}')
else:
  print('Data house tidak dicantumkan lengkap di PDF.')
```

```python
if 'house' in globals():
  model = sm.OLS(house[outcome], house[predictors].assign(const=1))
  results = model.fit()
  results.summary()
else:
  print('Data house tidak dicantumkan lengkap di PDF.')
```

**R-squared** mengukur proporsi variasi pada data yang dijelaskan oleh model:

$$R^2 = 1 - \frac{\sum_{i=1}^{n}(y_i - \hat{y_i})^2}{\sum_{i=1}^{n}(y_i - \bar{y})^2}$$

Software seperti R dan statsmodels juga menampilkan standard error, t-statistic, p-value, dan F-statistic. Dalam data science, t-statistic lebih sering dipakai sebagai petunjuk apakah predictor perlu dipertahankan atau dapat dihapus dari model.

### Cross-Validation

Metric klasik seperti R-squared, F-statistic, dan p-value adalah **in-sample metrics**, karena dihitung dari data yang sama dengan data training. Untuk menilai performa prediksi, sebagian data dapat disisihkan sebagai holdout.

**k-fold cross-validation** memperluas ide holdout dengan beberapa fold:

1. Sisihkan 1/k data sebagai holdout.
2. Latih model pada sisa data.
3. Terapkan model pada holdout dan catat metric.
4. Kembalikan fold pertama, lalu sisihkan fold berikutnya.
5. Ulangi sampai setiap record pernah menjadi holdout.
6. Gabungkan atau rata-ratakan metric penilaian model.

Pembagian data menjadi training sample dan holdout sample disebut **fold**.

### Model Selection and Stepwise Regression

Menambahkan variabel tidak selalu membuat model lebih baik. Prinsip **Occam's razor** digunakan sebagai panduan: jika hal lain setara, model yang lebih sederhana lebih disukai.

Menambahkan variabel selalu menurunkan RMSE dan menaikkan R-squared pada training data, sehingga metric tersebut tidak cukup untuk memilih model. Alternatifnya adalah adjusted R-squared atau AIC.

Adjusted R-squared:

$$R_{adj}^2 = 1 - (1 - R^2)\frac{n - 1}{n - P - 1}$$

AIC:

$$AIC = 2P + n\log(RSS/n)$$

Tujuannya adalah mencari model yang meminimalkan AIC.

```python
if 'house' in globals():
  predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms', 'BldgGrade',
                'PropertyType', 'NbrLivingUnits', 'SqFtFinBasement', 'YrBuilt',
                'YrRenovated', 'NewConstruction']

  X = pd.get_dummies(house[predictors], drop_first=True)
  X['NewConstruction'] = [1 if nc else 0 for nc in X['NewConstruction']]

  house_full = sm.OLS(house[outcome], X.assign(const=1))
  results = house_full.fit()
  results.summary()
else:
  print('Data house tidak dicantumkan lengkap di PDF.')
```

```python
if 'house' in globals() and 'X' in globals():
  try:
    from dmba import stepwise_selection, AIC_score

    y = house[outcome]

    def train_model(variables):
      if len(variables) == 0:
        return None
      model = LinearRegression()
      model.fit(X[variables], y)
      return model

    def score_model(model, variables):
      if len(variables) == 0:
        return AIC_score(y, [y.mean()] * len(y), model, df=1)
      return AIC_score(y, model.predict(X[variables]), model)

    best_model, best_variables = stepwise_selection(X.columns, train_model,
                                                    score_model, verbose=True)

    print(f'Intercept: {best_model.intercept_:.3f}')
    print('Coefficients:')
    for name, coef in zip(best_variables, best_model.coef_):
      print(f' {name}: {coef}')
  except ImportError:
    print('Package dmba tidak tersedia.')
else:
  print('Data house tidak dicantumkan lengkap di PDF.')
```

Stepwise regression dapat dilakukan dengan **backward elimination**, **forward selection**, atau kombinasi add/drop predictor untuk menurunkan AIC atau menaikkan adjusted R-squared.

**Penalized regression** memiliki semangat yang mirip AIC, tetapi penalty dimasukkan langsung ke persamaan fitting. Contohnya adalah **ridge regression** dan **lasso regression**.

## 4. Prediction Using Regression

Tujuan utama regression dalam data science adalah **prediction**.

| Istilah | Makna |
|---|---|
| Prediction interval | Interval uncertainty di sekitar satu nilai prediksi individual. |
| Extrapolation | Memperluas model di luar range data yang digunakan untuk fitting. |

### The Dangers of Extrapolation

Regression model tidak boleh digunakan untuk extrapolate di luar range data training. Model valid hanya untuk nilai predictor yang memiliki data cukup.

Contoh ekstrem dari materi: model rumah digunakan untuk memprediksi nilai tanah kosong 5.000 square foot. Karena data hanya berisi parcel dengan bangunan, model tidak memiliki informasi tentang vacant land dan menghasilkan prediksi negatif yang tidak masuk akal.

### Confidence and Prediction Intervals

Confidence interval mengukur uncertainty di sekitar regression coefficient atau statistik. Prediction interval mengukur uncertainty pada prediksi individual.

Bootstrap algorithm untuk confidence interval coefficient:

1. Anggap setiap baris data sebagai satu tiket.
2. Ambil tiket secara acak, catat nilainya, lalu kembalikan.
3. Ulangi sampai mendapatkan satu bootstrap resample berukuran n.
4. Fit regression pada bootstrap sample dan catat coefficient.
5. Ulangi, misalnya 1.000 kali.
6. Gunakan percentile coefficient untuk membentuk interval.

Untuk prediction interval, uncertainty berasal dari dua sumber: uncertainty model regression dan error individual data point. Prediction interval biasanya lebih lebar daripada confidence interval karena mengukur uncertainty pada satu nilai individual.

**Key Ideas:**

- Extrapolation di luar range data dapat menyebabkan error.
- Confidence interval mengukur uncertainty di sekitar coefficient regression.
- Prediction interval mengukur uncertainty pada prediksi individual.
- Bootstrap dapat digunakan untuk membentuk confidence interval dan prediction interval.

## 5. Factor Variables in Regression

Factor variables atau categorical variables memiliki sejumlah nilai diskret terbatas. Karena regression membutuhkan input numerik, factor variables perlu dikodekan ulang menjadi variabel numerik.

| Istilah | Makna |
|---|---|
| Dummy variables | Variabel binary 0-1 hasil recoding factor data. |
| Reference coding | Satu level factor menjadi reference, level lain dibandingkan terhadap reference tersebut. |
| One hot encoder | Semua level factor dipertahankan sebagai binary variables. Tidak sesuai untuk multiple linear regression jika intercept dipakai. |
| Deviation coding | Setiap level dibandingkan terhadap overall mean. |

Contoh materi menggunakan variable **PropertyType** dengan tiga nilai: **Multiplex**, **Single Family**, dan **Townhouse**.

```python
if 'house' in globals():
  house.PropertyType.head()
else:
  pd.Series(['Multiplex', 'Single Family', 'Single Family', 'Single Family', 'Single Family', 'Townhouse'],
            name='PropertyType')
```

Secara default, `pd.get_dummies` menghasilkan one hot encoding. Argumen `drop_first=True` menghasilkan **P - 1** columns. Ini digunakan untuk menghindari multicollinearity pada multiple linear regression dengan intercept.

```python
if 'house' in globals():
  predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms',
                'BldgGrade', 'PropertyType']

  X = pd.get_dummies(house[predictors], drop_first=True)

  house_lm_factor = LinearRegression()
  house_lm_factor.fit(X, house[outcome])

  print(f'Intercept: {house_lm_factor.intercept_:.3f}')
  print('Coefficients:')
  for name, coef in zip(X.columns, house_lm_factor.coef_):
    print(f' {name}: {coef}')
else:
  print('Data house tidak dicantumkan lengkap di PDF.')
```

Pada regression, factor variable dengan **P** level biasanya direpresentasikan dengan **P - 1** columns karena model memiliki intercept. Jika semua P dummy dimasukkan, satu column menjadi redundant dan dapat menyebabkan multicollinearity.

### Factor Variables with Many Levels

Factor variable bisa menghasilkan dummy variable sangat banyak. Contohnya adalah **ZipCode**. Pada King County, ada 80 zip code dengan penjualan rumah. ZipCode penting karena menjadi proxy lokasi, tetapi memasukkan semua level membutuhkan banyak coefficient dan degrees of freedom.

Alternatif yang digunakan dalam materi adalah mengelompokkan zip code berdasarkan median residual dari model awal, lalu membagi zip code menjadi lima **ZipGroup**.

```python
if 'house' in globals():
  pd.DataFrame(house['ZipCode'].value_counts()).transpose()
else:
  print('Data house tidak dicantumkan lengkap di PDF.')
```

```python
if 'house' in globals() and 'house_lm' in globals():
  zip_groups = pd.DataFrame([
    *pd.DataFrame({
      'ZipCode': house['ZipCode'],
      'residual': house[outcome] - house_lm.predict(house[predictors]),
    })
    .groupby(['ZipCode'])
    .apply(lambda x: {
      'ZipCode': x.iloc[0, 0],
      'count': len(x),
      'median_residual': x.residual.median()
    })
  ]).sort_values('median_residual')

  zip_groups['cum_count'] = np.cumsum(zip_groups['count'])
  zip_groups['ZipGroup'] = pd.qcut(zip_groups['cum_count'], 5, labels=False,
                                   retbins=False)
  to_join = zip_groups[['ZipCode', 'ZipGroup']].set_index('ZipCode')
  house = house.join(to_join, on='ZipCode')
  house['ZipGroup'] = house['ZipGroup'].astype('category')
else:
  print('Data house tidak dicantumkan lengkap di PDF.')
```

## 6. Interpreting the Regression Equation

Dalam data science, penggunaan utama regression adalah prediction. Namun, kadang persamaan regression juga digunakan untuk memahami hubungan predictor dan outcome.

| Istilah | Makna |
|---|---|
| Correlated variables | Predictor variables yang sangat berkorelasi sehingga coefficient sulit diinterpretasikan. |
| Multicollinearity | Predictor variables memiliki perfect atau near-perfect correlation sehingga regression tidak stabil atau tidak dapat dihitung. |
| Confounding variables | Predictor penting yang dihilangkan sehingga model menghasilkan hubungan palsu. |
| Main effects | Hubungan predictor dengan outcome secara independen dari variabel lain. |
| Interactions | Hubungan antar predictor dan response yang saling bergantung. |

### Correlated Predictors

Pada multiple regression, predictors sering berkorelasi satu sama lain. Dalam model rumah, coefficient **Bedrooms** dapat bernilai negatif. Ini terjadi karena rumah yang lebih besar cenderung punya lebih banyak kamar, dan ukuran rumah lebih mendorong nilai rumah daripada jumlah kamar.

```python
if 'best_model' in globals() and 'best_variables' in globals():
  print(f'Intercept: {best_model.intercept_:.3f}')
  print('Coefficients:')
  for name, coef in zip(best_variables, best_model.coef_):
    print(f' {name}: {coef}')
else:
  print('Model stepwise belum tersedia.')
```

```python
if 'house' in globals():
  predictors = ['Bedrooms', 'BldgGrade', 'PropertyType', 'YrBuilt']
  outcome = 'AdjSalePrice'

  X = pd.get_dummies(house[predictors], drop_first=True)

  reduced_lm = LinearRegression()
  reduced_lm.fit(X, house[outcome])
else:
  print('Data house tidak dicantumkan lengkap di PDF.')
```

Saat variabel **SqFtTotLiving**, **SqFtFinBasement**, dan **Bathrooms** dihapus, coefficient **Bedrooms** menjadi positif. Ini menunjukkan bahwa coefficient dapat berubah karena korelasi antar predictor.

### Multicollinearity

Multicollinearity terjadi ketika ada redundancy antar predictor variables. Contohnya:

- variabel yang sama dimasukkan lebih dari satu kali;
- dari factor variable dibuat P dummies, bukan P - 1;
- dua variabel hampir perfectly correlated.

Multicollinearity harus ditangani dengan menghapus variabel sampai redundancy hilang. Pada perfect multicollinearity, regression tidak memiliki solusi yang terdefinisi dengan baik.

### Confounding Variables

Confounding variable adalah variabel penting yang tidak dimasukkan ke dalam model. Pada data rumah, lokasi adalah predictor penting. Tanpa lokasi, coefficient **SqFtLot**, **Bathrooms**, dan **Bedrooms** dapat terlihat negatif.

Materi memasukkan **ZipGroup** untuk merepresentasikan lokasi.

```python
if 'house' in globals() and 'ZipGroup' in house.columns:
  predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms',
                'BldgGrade', 'PropertyType', 'ZipGroup']
  outcome = 'AdjSalePrice'

  X = pd.get_dummies(house[predictors], drop_first=True)

  confounding_lm = LinearRegression()
  confounding_lm.fit(X, house[outcome])

  print(f'Intercept: {confounding_lm.intercept_:.3f}')
  print('Coefficients:')
  for name, coef in zip(X.columns, confounding_lm.coef_):
    print(f' {name}: {coef}')
else:
  print('Data house atau ZipGroup tidak tersedia.')
```

Setelah **ZipGroup** dimasukkan, **SqFtLot** dan **Bathrooms** menjadi positif. Materi mencatat bahwa rumah di ZipGroup paling mahal diperkirakan memiliki sale price hampir **$340,000** lebih tinggi.

### Interactions and Main Effects

Main effects adalah predictor variables dalam regression equation. Jika hanya main effects dipakai, model mengasumsikan hubungan predictor dengan response tidak bergantung pada predictor lain.

Dalam data rumah, hubungan antara ukuran rumah dan harga dapat bergantung pada lokasi. Interaksi antara **SqFtTotLiving** dan **ZipGroup** digunakan untuk memodelkan hal ini.

```python
if 'house' in globals() and 'ZipGroup' in house.columns:
  model = smf.ols(formula='AdjSalePrice ~ SqFtTotLiving*ZipGroup + SqFtLot + ' +
                  'Bathrooms + Bedrooms + BldgGrade + PropertyType', data=house)
  results = model.fit()
  results.summary()
else:
  print('Data house atau ZipGroup tidak tersedia.')
```

Pada materi, location dan house size memiliki interaction yang kuat. Untuk ZipGroup paling mahal, slope dari SqFtTotLiving adalah jumlah main effect dan interaction term, sehingga tambahan satu square foot memberi peningkatan nilai yang jauh lebih besar daripada di ZipGroup rendah.

**Key Ideas:**

- Correlation antar predictors membuat coefficient multiple linear regression harus diinterpretasikan hati-hati.
- Multicollinearity dapat menyebabkan instability numerik dalam fitting regression.
- Confounding variable yang dihilangkan dapat menghasilkan hubungan palsu.
- Interaction term diperlukan ketika hubungan variabel dan response saling bergantung.

## 7. Regression Diagnostics

Regression diagnostics digunakan untuk menilai model, terutama melalui analisis residuals. Pada data science, diagnostics tidak selalu langsung menilai predictive accuracy, tetapi dapat memberi insight untuk setting prediksi.

| Istilah | Makna |
|---|---|
| Standardized residuals | Residuals dibagi standard error residuals. |
| Outliers | Record atau outcome yang jauh dari data lain atau prediksi. |
| Influential value | Record yang kehadiran atau ketiadaannya membuat regression equation berubah besar. |
| Leverage | Tingkat pengaruh satu record terhadap regression equation. |
| Non-normal residuals | Residuals yang tidak normal; biasanya bukan perhatian utama data science. |
| Heteroskedasticity | Variance residual berbeda pada beberapa range outcome. |
| Partial residual plots | Plot diagnostic untuk melihat hubungan outcome dengan satu predictor. |

### Outliers

Dalam regression, outlier adalah record dengan nilai aktual **y** yang jauh dari nilai prediksi. Outlier dapat dideteksi dengan **standardized residual**, yaitu residual dibagi standard error residual.

Materi memberi contoh fitting regression untuk rumah pada zip code **98105**.

```python
if 'house' in globals():
  house_98105 = house.loc[house['ZipCode'] == 98105, ]

  predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms', 'BldgGrade']
  outcome = 'AdjSalePrice'

  house_outlier = sm.OLS(house_98105[outcome],
                         house_98105[predictors].assign(const=1))
  result_98105 = house_outlier.fit()
else:
  print('Data house tidak dicantumkan lengkap di PDF.')
```

```python
if 'result_98105' in globals():
  influence = OLSInfluence(result_98105)
  sresiduals = influence.resid_studentized_internal
  sresiduals.idxmin(), sresiduals.min()
else:
  print('Model result_98105 belum tersedia.')
```

```python
if 'house_98105' in globals() and 'sresiduals' in globals():
  outlier = house_98105.loc[sresiduals.idxmin(), :]
  print('AdjSalePrice', outlier[outcome])
  print(outlier[predictors])
else:
  print('Data house_98105 atau residuals belum tersedia.')
```

Pada materi, outlier terbesar adalah overestimate lebih dari empat standard errors dari regression line. Record tersebut tampak tidak normal karena rumah dengan ukuran tersebut biasanya dijual jauh lebih mahal pada zip code tersebut. Materi menyebut bahwa penjualan itu hanya melibatkan partial interest pada properti.

### Influential Values

Influential observation adalah record yang jika dihapus akan mengubah regression equation secara besar. Nilai seperti ini tidak harus memiliki residual besar. Dalam regression, pengaruh ini disebut **leverage**.

Metric yang digunakan antara lain **hat-value** dan **Cook's distance**. Influence plot menggabungkan standardized residuals, hat-value, dan Cook's distance.

```python
if 'result_98105' in globals():
  influence = OLSInfluence(result_98105)
  fig, ax = plt.subplots(figsize=(5, 5))
  ax.axhline(-2.5, linestyle='--', color='C1')
  ax.axhline(2.5, linestyle='--', color='C1')
  ax.scatter(influence.hat_matrix_diag, influence.resid_studentized_internal,
             s=1000 * np.sqrt(influence.cooks_distance[0]),
             alpha=0.5)
  ax.set_xlabel('hat values')
  ax.set_ylabel('studentized residuals')
else:
  print('Model result_98105 belum tersedia.')
```

Materi membandingkan coefficient regression sebelum dan sesudah data highly influential dihapus.

```python
pd.DataFrame({
  'Original': [-772550, 210, 39, 2282, -26320, 130000],
  'Influential removed': [-647137, 230, 33, -16132, -22888, 114871]
}, index=['(Intercept)', 'SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms', 'BldgGrade'])
```

Pada data besar, influential observation biasanya tidak terlalu bermasalah untuk fitting prediktif karena satu record sulit mengubah model secara ekstrem. Namun, untuk anomaly detection, influential observations dapat sangat berguna.

### Heteroskedasticity, Non-Normality, and Correlated Errors

Heteroskedasticity adalah kondisi saat variance residual tidak konstan di seluruh range predicted values. Artinya, error lebih besar pada bagian tertentu dari range.

Dalam data science, residual distribution biasanya bukan perhatian utama jika fokusnya predictive accuracy. Namun, heteroskedasticity dapat menunjukkan bahwa model belum menangkap sinyal tertentu.

```python
if 'result_98105' in globals():
  fig, ax = plt.subplots(figsize=(5, 5))
  sns.regplot(x=result_98105.fittedvalues, y=np.abs(result_98105.resid),
              scatter_kws={'alpha': 0.25}, line_kws={'color': 'C1'},
              lowess=True, ax=ax)
  ax.set_xlabel('predicted')
  ax.set_ylabel('abs(residual)')
else:
  print('Model result_98105 belum tersedia.')
```

Materi juga menyebut histogram standardized residuals untuk regression **lm_98105**. Residuals memiliki tail lebih panjang daripada normal distribution dan mild skewness ke residual yang lebih besar.

Statisticians juga dapat memeriksa independence error, terutama untuk data time series atau spatial. Durbin-Watson statistic dapat digunakan untuk mendeteksi autocorrelation pada regression untuk time series.

### Partial Residual Plots and Nonlinearity

Partial residual plot digunakan untuk melihat seberapa baik estimated fit menjelaskan hubungan antara satu predictor dan outcome, dengan mempertimbangkan predictor lainnya.

Partial residual untuk predictor **Xi**:

$$Partial\ residual = Residual + b_iX_i$$

```python
if 'result_98105' in globals():
  sm.graphics.plot_ccpr(result_98105, 'SqFtTotLiving')
else:
  print('Model result_98105 belum tersedia.')
```

Pada materi, partial residual plot menunjukkan hubungan **SqFtTotLiving** dan sales price yang nonlinear. Regression line meng-underestimate harga untuk rumah kurang dari 1.000 square feet dan meng-overestimate harga untuk rumah antara 2.000 sampai 3.000 square feet.

**Key Ideas:**

- Outlier dalam regression adalah record dengan residual besar.
- Fokus utama outlier adalah menemukan masalah data atau anomali.
- Satu record dapat sangat memengaruhi regression pada data kecil, tetapi efek ini biasanya berkurang pada data besar.
- Distribution residual tidak kritis untuk data science jika fokusnya predictive accuracy.
- Partial residual plot dapat digunakan untuk menilai fit pada setiap regression term dan mengarah pada specification model alternatif.

## 8. Polynomial and Spline Regression

Hubungan antara response dan predictor tidak selalu linear. Response terhadap dosis obat, misalnya, sering nonlinear. Regression dapat diperluas agar memasukkan hubungan nonlinear.

| Istilah | Makna |
|---|---|
| Polynomial regression | Menambahkan polynomial terms seperti square atau cube ke regression. |
| Spline regression | Memasang smooth curve dengan serangkaian polynomial segments. |
| Knots | Nilai yang memisahkan spline segments. |
| Generalized additive models | Spline model dengan pemilihan knots otomatis; disebut juga GAM. |

### Polynomial

Polynomial regression memasukkan polynomial terms ke dalam regression equation. Contoh quadratic regression:

$$Y = b_0 + b_1X + b_2X^2 + e$$

Materi menggunakan quadratic polynomial untuk **SqFtTotLiving**.

```python
if 'house_98105' in globals():
  model_poly = smf.ols(formula='AdjSalePrice ~ SqFtTotLiving + ' +
                       '+ I(SqFtTotLiving**2) + ' +
                       'SqFtLot + Bathrooms + Bedrooms + BldgGrade', data=house_98105)
  result_poly = model_poly.fit()
  result_poly.summary()
else:
  print('Data house_98105 tidak tersedia.')
```

Pada model polynomial, ada dua coefficient untuk **SqFtTotLiving**: satu untuk linear term dan satu untuk quadratic term.

### Splines

Polynomial regression hanya menangkap sejumlah curvature tertentu. Polynomial orde tinggi dapat menghasilkan bentuk yang terlalu bergelombang. Alternatifnya adalah **splines**.

Spline adalah rangkaian piecewise continuous polynomials yang disambungkan halus pada titik tetap yang disebut **knots**.

Dalam R, materi memakai `bs` dari package `splines`. Dalam Python, statsmodels formula interface mendukung spline dengan `bs`.

```python
if 'house_98105' in globals():
  formula = 'AdjSalePrice ~ bs(SqFtTotLiving, df=6, degree=3) + ' + \
            'SqFtLot + Bathrooms + Bedrooms + BldgGrade'
  model_spline = smf.ols(formula=formula, data=house_98105)
  result_spline = model_spline.fit()
else:
  print('Data house_98105 tidak tersedia.')
```

Coefficient pada spline term tidak diinterpretasikan langsung seperti linear term. Visual display lebih berguna untuk melihat bentuk spline fit.

### Generalized Additive Models

Jika hubungan nonlinear dicurigai, polynomial terms mungkin tidak cukup fleksibel, dan spline terms membutuhkan penentuan knots. **Generalized additive models (GAM)** dapat digunakan untuk otomatis fitting spline regression.

Dalam Python, materi menggunakan package **pyGAM** dan class **LinearGAM**.

```python
if 'house_98105' in globals():
  try:
    from pygam import LinearGAM, s, l

    predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms', 'BldgGrade']
    outcome = 'AdjSalePrice'
    X = house_98105[predictors].values
    y = house_98105[outcome]

    gam = LinearGAM(s(0, n_splines=12) + l(1) + l(2) + l(3) + l(4))
    gam.gridsearch(X, y)
  except ImportError:
    print('Package pyGAM tidak tersedia.')
else:
  print('Data house_98105 tidak tersedia.')
```

Materi menyebut default `n_splines` adalah 20 dan dapat menyebabkan overfitting untuk nilai **SqFtTotLiving** yang lebih besar. Nilai 12 menghasilkan fit yang lebih masuk akal.

**Key Ideas:**

- Outlier dalam regression adalah record dengan residual besar.
- Multicollinearity dapat menyebabkan instability dalam fitting regression equation.
- Confounding variable yang dihilangkan dapat menghasilkan hubungan palsu.
- Interaction term dibutuhkan jika efek satu variabel bergantung pada level atau magnitude variabel lain.
- Polynomial regression dapat memodelkan hubungan nonlinear antara predictors dan outcome.
- Splines adalah rangkaian polynomial segments yang bergabung pada knots.
- GAM dapat mengotomatisasi proses spesifikasi knots pada splines.

## 9. Summary

Regression adalah proses membangun hubungan antara beberapa predictor variables dan outcome variable. Bentuk dasarnya linear: setiap predictor memiliki coefficient yang menjelaskan hubungan linear dengan outcome. Bentuk yang lebih lanjut, seperti polynomial regression dan spline regression, memungkinkan hubungan nonlinear.

Dalam statistik klasik, fokus regression adalah mendapatkan fit yang baik terhadap data yang diamati untuk menjelaskan fenomena. Dalam data science, tujuannya biasanya memprediksi nilai untuk data baru, sehingga metric yang digunakan lebih fokus pada predictive accuracy untuk out-of-sample data. Variable selection digunakan untuk mengurangi dimensionality dan membuat model lebih compact.
